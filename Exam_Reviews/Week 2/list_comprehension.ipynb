```python


def check_palindrome(number):
    original_number = number
    reverse = 0
    while number>0:
        digit = number%10
        reverse = reverse *10 + digit
        number = number // 10

    if original_number == reverse:
        print("Palindrome")
    else:
        print("Not Palindrome")


number = int(input("Enter a Number:"))

check_palindrome(number)
```
