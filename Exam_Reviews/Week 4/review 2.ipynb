````python
# ractical Task: Statistical Analysis for ML Model Performance

# You are given the following dataset of study hours and exam scores for 10 students.

# ```
# import numpy as np import pandas as pd import matplotlib.pyplot as pltfrom scipy import stats data = pd.DataFrame({    "study_hours": [1, 2, 2, 3, 4, 5, 6, 7, 8, 10],    "exam_score": [35, 40, 42, 45, 55, 60, 65, 72, 80, 95]})
# ```

# Requirements

# 1. Calculate mean, median, mode, variance, standard deviation, and range of exam scores.
# 2. Calculate Pearson and Spearman correlations between study hours and exam scores.
# 3. Plot a histogram and box plot of exam scores.
# 4. Assuming exam scores are normally distributed, calculate the probability of a score exceeding 70 using a normal distribution CDF.
# 5. Conduct a one-sample t-test to determine whether the mean exam score differs significantly from 50, using a significance level of 0.05.
# 6. Interpret whether study hours are associated with exam performance and explain why correlation does not establish causation.

# Constraint: Use Python, NumPy, Pandas, SciPy, and Matplotlib. Provide numerical results and a brief statistical conclusion.
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

data = pd.DataFrame({
    "study_hours": [1, 2, 2, 3, 4, 5, 6, 7, 8, 10],
    "exam_score": [35, 40, 42, 45, 55, 60, 65, 72, 80, 95]
})

data
````

````python
scores = data["exam_score"]

mean = scores.mean()
median = scores.median()
mode = scores.mode().tolist()
variance = scores.var(ddof=1)
std_dev = scores.std(ddof=1)
score_range = scores.max() - scores.min()

print("Mean:", round(mean, 2))
print("Median:", round(median, 2))
print("Mode:", mode)
print("Sample Variance:", round(variance, 2))
print("Sample Standard Deviation:", round(std_dev, 2))
print("Range:", score_range)
````

````python
pearson_r, pearson_p = stats.pearsonr(
    data["study_hours"],
    data["exam_score"]
)

spearman_r, spearman_p = stats.spearmanr(
    data["study_hours"],
    data["exam_score"]
)

print("Pearson Correlation:", round(pearson_r, 4))
print("Pearson p-value:", pearson_p)

print("Spearman Correlation:", round(spearman_r, 4))
print("Spearman p-value:", spearman_p)
````

````python
plt.hist(data["exam_score"], bins=5, edgecolor="black")

plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")

plt.show()
````

````python
plt.boxplot(data["exam_score"])

plt.ylabel("Exam Score")
plt.title("Box Plot of Exam Scores")

plt.show()
````

````python
mean = data["exam_score"].mean()
std_dev = data["exam_score"].std(ddof=1)

probability = 1 - stats.norm.cdf(
    70,
    loc=mean,
    scale=std_dev
)

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std_dev, 2))
print("Probability of scoring above 70:", round(probability, 4))
print("Percentage:", round(probability * 100, 2), "%")
````

````python
scores = data["exam_score"]

t_statistic, p_value = stats.ttest_1samp(scores, popmean=50)

alpha = 0.05

print("Sample Mean:", round(scores.mean(), 2))
print("T-statistic:", round(t_statistic, 4))
print("P-value:", round(p_value, 4))

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("The mean score differs significantly from 50.")
else:
    print("Fail to reject the null hypothesis.")
    print("The mean score does not differ significantly from 50.")
````
