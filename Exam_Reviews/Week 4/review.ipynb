````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
````

````python
data = pd.DataFrame({    
        "study_hours": [1, 2, 2, 3, 4, 5, 6, 7, 8, 10],
        "exam_score": [35, 40, 42, 45, 55, 60, 65, 72, 80, 95]
})

data
````

````python
exam_score = data["exam_score"]
n = len(exam_score)

mean = sum(exam_score)/n
median = np.median(exam_score)
variance = sum([(x-mean)**2 for x in exam_score])
std_dev = variance ** 0.5
range = max(exam_score) - min(exam_score)
variance
````

````python
pearson_corr = data.corr(method="pearson")
spearman_corr = data.corr(method="spearman")
````

````python
plt.hist(exam_score, bins=5,
         rwidth=0.90)
plt.title("Exam Score Distribution")
plt.xlabel("Exam Scores")
plt.ylabel("Count")

plt.show()
````

````python
# Assuming exam scores are normally distributed, calculate the probability of a score exceeding 70 using a normal distribution CDF.
````

````python
exam_score 
exam_sc = np.asarray(exam_score)
exam_sc

event = [x for x in exam_sc if x>70 ]

prob_g_70 = len(event) / len(exam_score)
print(prob_g_70 * 100)
````
