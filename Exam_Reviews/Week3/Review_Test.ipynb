```python
import pandas as pd


pokemone_df = pd.read_csv("pokemone.csv")
pokemone_df.info()
```

```python
pokemone_df["Weight"] = pokemone_df["Weight"].astype(int)
pokemone_df.info()
pokemone_df
```

```python
pokemone_df["No"].remove()
```

```python
avg_type1 = pokemone_df.groupby("Type1")["Weight"].mean()
print(avg_type1)
```
