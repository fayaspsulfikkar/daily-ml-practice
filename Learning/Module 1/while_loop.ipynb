```python
#Sum 1 to N
n=int(input("Enter a number: "))
i=1
tot=0
while i<=n:
    tot+=i
    i +=1
print(tot)
```

```python
#Multiplication table

num=int(input("Enter a Number"))
i=1

while i<11:
    mul=i*num
    print(i,"*",num,"=",mul)
    i +=1
```

```python
#Count digits
num=int(input("Enter a Number: "))
count=0

while num>0:
    num=num//10
    count +=1
print(count)
```
