```python
# Exercise 1: locate the shared data folder from this notebook folder
from pathlib import Path
import csv
import json

current_folder = Path.cwd().resolve()
project_root = next(folder for folder in [current_folder, *current_folder.parents] if (folder / "data").is_dir())
data_folder = project_root / "data"
print(data_folder)
```

```python
# Exercise 2: write and read a plain-text note
note_path = data_folder / "practice_note.txt"
note_path.write_text("Python file handling practice\n", encoding="utf-8")
print(note_path.read_text(encoding="utf-8"))
```

```python
# Exercise 3: read CSV rows as dictionaries
student_path = data_folder / "students.csv"
with student_path.open(newline="", encoding="utf-8") as file:
    students = list(csv.DictReader(file))
print(students[0])
```

```python
# Exercise 4: calculate an average from CSV data
scores = [int(student["score"]) for student in students]
print("Average score:", sum(scores) / len(scores))
```

```python
# Exercise 5: load and inspect JSON data
profile_path = data_folder / "profile.json"
with profile_path.open(encoding="utf-8") as file:
    profile = json.load(file)
print(profile["name"], "is learning", ", ".join(profile["skills"]))
```

```python
# Practice challenge: save a filtered CSV report
passed_students = [student for student in students if int(student["score"]) >= 75]
report_path = data_folder / "passed_students.csv"
with report_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "score", "city"])
    writer.writeheader()
    writer.writerows(passed_students)
print("Created:", report_path.name)
```
