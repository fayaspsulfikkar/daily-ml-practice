```python
# Exercise 2: reproduce and fix a likely bug
def safe_average(values):
    if not values:
        return 0
    return sum(values) / len(values)


print(safe_average([]))
print(safe_average([50, 75, 100]))
```

```python
# Exercise 3: compare two implementations with the same tests
def count_even_loop(numbers):
    count = 0
    for number in numbers:
        if number % 2 == 0:
            count += 1
    return count


def count_even_comprehension(numbers):
    return len([number for number in numbers if number % 2 == 0])


test_values = [1, 2, 3, 4, 5, 6]
print(count_even_loop(test_values), count_even_comprehension(test_values))
```

```python
# Exercise 4: use tests to validate a generated or refactored function
def title_case_name(name):
    return " ".join(part.capitalize() for part in name.split())


assert title_case_name("  aSHA  naIR ") == "Asha Nair"
assert title_case_name("") == ""
print("All checks passed")
```
