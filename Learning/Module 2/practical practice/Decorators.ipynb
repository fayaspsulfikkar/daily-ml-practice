```python
def divide(a,b):
    if a<b:
        a,b = b,a
    return a/b

result = divide(2,4)

print(result)
```

```python
#what_if_there_are_multiple_functions_that_needs_same_line_of_code

def divide(a,b):
    if a<b:
        a,b = b,a
    return a/b
def sub(a,b):
    if a<b:
        a,b = b,a
    return a-b

result1 = sub(2,4)
print(result1)
```

```python
#instead_use_a_decorator

def greater_first(func):
    def wrap(a,b):
        if a<b:
            a,b = b,a
        return func(a,b)
    return wrap

@greater_first
def divide(a,b):
    return a/b

@greater_first
def sub(a,b):
    return a-b

result1 = sub(2,4)
print(result1)
```
