```python
#Check whether a number is prime

number = int(input("Enter a number: "))

if number <= 1:
    print("The Number is not Prime")
else:

    prime = True
    for num in range(2,number):
        if number%num == 0:
            prime = False
            break
    if prime == False:
        print("Number is not prime")
    else:
        print("Number is prime")
```
