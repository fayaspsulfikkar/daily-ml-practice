```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
categories = ["Grains","Vegetables","Fruits","Protein","Diary","Sugar"]
consumption = [1,4,2,5,3,1]

plt.grid(axis="both",
         linestyle="dashdot",
         linewidth="0.5",
         color="grey")

plt.bar(categories,consumption,
        color="red"
        )
plt.tick_params(axis="both",
                color="blue")
plt.title("Food Intake",
          fontsize=20,
          fontweight="bold",
          family="Sans",
          color="Blue"
          )
plt.xticks(categories,
           color="black")
plt.yticks(consumption,
           color="black")
plt.xlabel("Foods",
           color="blue",
           )
plt.ylabel("Consumption",
           color="blue")

plt.show()
```
