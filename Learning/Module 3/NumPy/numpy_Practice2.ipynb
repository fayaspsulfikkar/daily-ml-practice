```python
import numpy as np

salaries = np.array([25000, 42000, 18000, 55000, 32000, 60000, 15000])
```

```python
#Increase every salary by 10% using vectorized computation.

new_sal = salaries + (salaries * 0.10)

print(new_sal)
```

```python
#Convert the updated salaries into integers.

int_new_sal = new_sal.astype(int)

print(int_new_sal)
print(int_new_sal.dtype)
```

```python
#Classify each updated salary as "High" if it is greater than or equal to ₹40,000; otherwise, "Low".

print(np.where(int_new_sal>=40000,"High","Low"  ))
```
