```python
import numpy as np
import time

numbers = list(range(1_000_000))
arr = np.array(numbers)

start = time.time()
list_result = [x + 5 for x in numbers]
list_time = time.time() - start

start = time.time()
array_result = arr + 5
numpy_time = time.time() - start

print(f"Python List: {list_time:.6f} seconds")
print(f"NumPy Array: {numpy_time:.6f} seconds")
print(f"Time gap: {list_time - numpy_time:.6f} seconds")
```

```python
arr1 = np.array([1, 4, 9, 16, 25])
arr2 = np.array([0, 1, 2, 3, 4])
arr3 = np.array([1, 2, 3, 4, 5])
arr4 = np.array([-10, -5, 0, 5, 10])
```
