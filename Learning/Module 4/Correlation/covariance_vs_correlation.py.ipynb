```python
# Goal: Understand the difference between covariance and correlation.
# Data: x = [1, 2, 3, 4], y = [2, 4, 6, 8]
# Tasks
# (a) Determine whether x and y tend to move in the same direction or opposite directions. Explain why.
# (b) Create a Pandas DataFrame using x and y. Calculate the covariance between the two variables using Pandas.
# (c) Calculate the Pearson correlation coefficient between x and y using Pandas.
# (d) Explain why covariance depends on the measurement units of the variables, while correlation is standardised and ranges from -1 to +1.
```

```python
import pandas as pd

```

```python
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

dict = {
    "x": x,
    "y": y
}
```

```python
df = pd.DataFrame(dict)

cova = df.cov()
cova
```

```python
corre = df.corr()
corre
```
