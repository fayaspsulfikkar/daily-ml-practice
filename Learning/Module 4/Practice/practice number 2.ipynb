```python
# Q11
# Calculate the population covariance using Python.
# x = [2, 4, 6, 8]y = [3, 6, 9, 12]
```

```python
# Q15
# Calculate Pearson correlation using:
# x = [2, 4, 6, 8]y = [4, 8, 12, 16]
# Calculate the means, deviations, population covariance, both population standard deviations, and finally the correlation.
```

```python
x = [2, 4, 6, 8]
y = [4, 8, 12, 16]

n = len(x)

mean_x = sum(x)/n
mean_y = sum(y)/n

devi_x = [(a-mean_x) for a in x]
devi_y = [(b-mean_y) for b in y]

products = [x*y for x,y in zip(devi_x,devi_y)]

pop_covvv = sum(products)/n


var_x = sum([(c**2) for c in devi_x])/n
var_y = sum([(d**2) for d in devi_y])/n

std_x = var_x ** 0.5 
std_y = var_y ** 0.5 

correlation = pop_covvv / (std_x * std_y)

print(correlation)
```

```python
# Q16
# Calculate Pearson correlation:
# temperature = [10, 20, 30, 40]sales = [40, 30, 20, 10]
# Use the same approach. Don't change your method.
```

```python
temperature = [10, 20, 30, 40]
sales = [40, 30, 20, 10]
n = len(temperature)

mean_temp = sum(temperature)/n
mean_sales = sum(sales)/n

dev_temp = [(x-mean_temp) for x in temperature]
dev_sale = [(y-mean_sales) for y in sales]

products = [x*y for x,y in zip(dev_temp,dev_sale)]

covv = sum(products)/n

var_temp = sum([(x**2) for x in dev_temp])/n
var_sale = sum([(x**2) for x in dev_sale])/n


std_temp = var_temp ** 0.5
std_sale = var_sale ** 0.5

corr = covv / (std_temp*std_sale)
print(corr)
```

```python
# Calculate Pearson correlation:
# study_hours = [1, 2, 3, 4, 5]scores = [40, 45, 55, 65, 70]
```

```python
study_hours = [1, 2, 3, 4, 5]
scores = [40, 45, 55, 65, 70]

n = len(study_hours)

mean_sh = sum(study_hours)/n
mean_sc = sum(scores)/n
print("Mean of Study Hours: ",mean_sh)
print("Mean of Scores: ",mean_sc)

dev_sh = [(x-mean_sh) for x in study_hours]
dev_sc = [(y-mean_sc) for y in scores]
print("Deviation of Study: ",dev_sh)
print("Deviation of Scores: ",dev_sc)

products = [x*y for x,y in zip(dev_sh,dev_sc)]

covariance = sum(products)/n

var_sh = sum([(x**2) for x in dev_sh])/n
var_sc = sum([(x**2) for x in dev_sc])/n

std_sh = var_sh ** 0.5 
std_sc = var_sc ** 0.5

correlation = covariance / (std_sh * std_sc)
print(correlation)
```

```python
# Q19
# x = [1, 2, 3, 4, 5]y = [5, 5, 5, 5, 5]
# Calculate the means, variances, and standard deviations. Then determine whether Pearson correlation can be calculated.
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 5, 5, 5, 5]

n = len(x)

meanx = sum(x)/n
meany = sum(y)/n

devx = [(a-meanx) for a in x]
devy = [(b-meany) for b in y]

product = [a*b for a,b in zip(devx,devy)]

covarian = sum(product)/n

varx = sum([a**2 for a in devx])/n
vary = sum([b**2 for b in devy])/n

stddv_x = varx ** 0.5
stddv_y = vary ** 0.5

if stddv_x == 0 or stddv_y ==0:
    print("Correlation is undefinded")
else:
    correla = covarian / (stddv_x * stddv_y)
    print(correla)
```

```python
# Q19
# Write a Python program to calculate all six:
# 1. Mean of both variables
# 2. Deviations of both variables
# 3. Population variance of both variables
# 4. Population standard deviation of both variables
# 5. Population covariance
# 6. Pearson correlation
# Use these datasets:
# temperature = [22, 31, 19, 35, 27]distance = [14, 8, 18, 5, 11]
```

```python
temperature = [22, 31, 19, 35, 27]
distance = [14, 8, 18, 5, 11]

n = len(temperature)

mean_te = sum(temperature)/n
mean_di = sum(distance)/n

dev_te = [(x-mean_te) for x in temperature]
dev_di = [(x-mean_di) for x in distance]

var_te = sum([a**2 for a in dev_te])/n
var_di = sum([b**2 for b in dev_di])/n

std_dev_te = var_te ** 0.5
std_dev_dis = var_di ** 0.5

products = [x*y for x,y in zip(dev_te,dev_di)]

pop_cova = sum(products)/n

correlation = pop_cova / (std_dev_te * std_dev_dis)

print(correlation)
```

```python
# Q20 — Final Challenge 🧠
# You have practised variance, standard deviation, covariance, and correlation. Now apply them to a real-world situation.

# Scenario: A student wants to understand whether studying more hours is associated with higher exam scores.
# study_hours = [2, 4, 1, 5, 3, 6]exam_scores = [45, 65, 40, 80, 50, 90]


# Your task
# Write a Python program that:
# 1. Calculates the mean of both variables.
# 2. Calculates the deviations from both means.
# 3. Calculates the population variance of both variables.
# 4. Calculates the population standard deviation of both variables.
# 5. Calculates the population covariance.
# 6. Calculates the Pearson correlation coefficient.
# 7. Interprets the correlation: Is the relationship positive or negative? Is it weak or strong?
```

```python
study_hours = [2, 4, 1, 5, 3, 6]
exam_scores = [45, 65, 40, 80, 50, 90]

n = len(study_hours)

mean_h = sum(study_hours) / n
mean_e = sum(exam_scores) / n

dev_h = [(x-mean_h) for x in study_hours]
dev_e = [(y-mean_e) for y in exam_scores]

vari_h = sum([x**2 for x in dev_h])/n
vari_e = sum([y**2 for y in dev_e])/n

std_dv_h = vari_h ** 0.5
std_dv_e = vari_e ** 0.5

products = [x*y for x,y in zip(dev_h,dev_e)]

popu_cova = sum(products)/n

corre = popu_cova / (std_dv_h * std_dv_e)
print(corre)

```
