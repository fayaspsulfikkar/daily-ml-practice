```python
# ### Probability — Random experiments, sample spaces, and events

# - **Goal:** describe possible outcomes and calculate an event probability.
# - **Data:** roll a fair six-sided die once.
# - **Steps:**
#     - (a) Write the sample space (all possible outcomes).
#     - (b) Define event A = “roll an even number.”
#     - (c) Calculate P(A).
#     - (d) Define event B = “roll a number greater than 4” and calculate P(B).
```

```python
sample_space = [1,2,3,4,5,6]
print("Sample Space: ",sample_space)
event_A = {x for x in sample_space if x%2==0}

p_a = len(event_A)/len(sample_space)
print("P(A): ",p_a)

event_B = {x for x in sample_space if x>4}
p_b = len(event_B)/len(sample_space)
print("P(B): ",p_b)
```
