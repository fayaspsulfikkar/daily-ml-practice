```python
# Imagine tossing a fair coin twice.

# Your task:
# 1. Create a set representing all possible outcomes of two tosses: HH, HT, TH, TT.
# 2. Find the probability of getting Heads on the first toss.
# 3. Find the probability of getting Heads on the second toss, given that the first toss was Heads.
```

```python
sample_s = {"HH","HT","TT","TH"}

A = {x for x in sample_s if x[0]=="H"}

P_A = len(A)/len(sample_s)

print("Outcomes: ",A)
print("Probability: ",P_A)
```
