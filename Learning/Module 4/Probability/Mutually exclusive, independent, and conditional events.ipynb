```python
# - **Goal:** distinguish three ideas that are often confused.
# - **Data:** roll one fair die. Let A = “roll 2” and B = “roll 5.”
# - **Steps:**
#     - (a) Can A and B happen on the same roll? Are they mutually exclusive?
#     - (b) Calculate P(A ∩ B).
#     - (c) For two separate fair coin tosses, explain whether the tosses are independent.
#     - (d) If you know a die roll was even, what is the probability it was 2?
```

```python
sample_s = {1,2,3,4,5,6}
A = {2}
B = {5}

a_b_inter = A & B

P_ab_inter = len(a_b_inter)/len(sample_s)
print(P_ab_inter)
```
