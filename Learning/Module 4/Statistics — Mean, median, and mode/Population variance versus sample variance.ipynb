```python
# - **Goal:** understand why two variance formulas divide by different values.
# - **Data:** values = [4, 6, 8]. Their mean is 6.
# - **Steps:**
#     - (a) Calculate the squared deviations from the mean.
#     - (b) Calculate population variance by dividing the squared-deviation total by n.
#     - (c) Calculate sample variance by dividing by n - 1.
#     - (d) Explain when each formula is appropriate.
```

```python
values = [4,6,8]

mean = sum(values)/len(values)

squared_deviation = []

for x in values:
    squared_deviation.append((x-mean)**2)

sumof = sum(squared_deviation)

population_variance = sumof / len(values)
population_variance

sample_variance = sumof/(len(values)-1)
sample_variance
```
