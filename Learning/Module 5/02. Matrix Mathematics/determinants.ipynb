```python
import numpy as np

A = np.array([
    [4, 2],
    [3, 5]
])

B = np.array([
    [1, 2, 3],
    [0, 4, 5],
    [1, 0, 6]
])

C = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [1, 0, 1]
])
```

```python
det_A = np.linalg.det(A)
det_B = np.linalg.det(B)
det_C = np.linalg.det(C)
print(det_A)
print(det_B)
print(det_C)
```

```python
print(det_C)
```

```python
B_swapped = np.array([[0, 4, 5],
                     [1, 2, 3],
                     [1, 0, 6]])
print(np.linalg.det(B_swapped))
```

```python
print(det_B == -np.linalg.det(B_swapped))
```

```python
B_swapped2 = np.array([
    [1, 0, 6],  
    [0, 4, 5], 
    [1, 2, 3]  
])
```

```python
print(np.linalg.det(B_swapped2))
```

```python
print(det_B == np.linalg.det(B_swapped2))
print(det_B,"\n",B_swapped2)
```

```python
print(np.isclose(det_B, -np.linalg.det(B_swapped2)))
```
