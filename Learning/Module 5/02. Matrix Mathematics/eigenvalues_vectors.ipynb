```python
import numpy as np 
```

```python
A = np.array([[4, 1],
              [2, 3]])

B = np.array([[5, 0],
              [0, 2]])
```

```python
trace_A = np.linalg.trace(A)
trace_B = np.linalg.trace(B)

det_A = np.linalg.det(A)
det_B = np.linalg.det(B)

print(trace_A)
print(trace_B)
print(det_A)
print(det_B)
```

```python
eigenvaluesA, eigenvectorsA = np.linalg.eig(A)
eigenvalues_B, eigenvectorsB = np.linalg.eig(B)

print(eigenvaluesA)
print(eigenvectorsA)
print(eigenvalues_B)
print(eigenvectorsB)
```
