```python
# Convert 30 degrees into radians using NumPy.
```

```python
angle = np.radians(30)
print(angle)
```

```python
angle_sin = np.sin(angle)
angle_cos = np.cos(angle)

print(angle_sin)
print(angle_cos)
```

```python
R = np.array([
    [angle_cos,angle_sin*-1],
    [angle_sin,angle_cos]
])
```

```python
R @ R.T
```

```python
np.allclose(R @ R.T, np.eye(2))
```

```python
print(R)
```

```python
v = np.array([3, 4])
rotated_v = v @ R
print(rotated_v)
```

```python
print(np.linalg.norm(v))
print(np.linalg.norm(rotated_v))
```
