```python
import numpy as np
```

```python
x = [1, 2, 3]
y = [2, 4, 6]

m = len(x)
```

```python
def gradient(w):
    total = 0
    for i in range(m):
        total += (w*x[i] - y[i])* x[i]

    total = total/m

    return total
```

```python
print(gradient(5))
```

```python
w = 0 
alpha = 0.1 


for i in range(2):
    grad = gradient(w)
    w = w - alpha * grad

    print(f"Iteration {i+1}: w = {w}")

```
