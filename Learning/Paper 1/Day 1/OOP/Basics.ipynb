```python
class Computer:

    def __init__(self,cpu,ram,ssd):
        self.cpu = cpu 
        self.ram = ram
        self.ssd = ssd


    def config(self):
        print("Config: ",self.cpu, self.ram, self.ssd)


com1 = Computer("i5","16 GB","512")
com2 = Computer("i7","32 GB", "1 TB")

Computer.config(com1)
Computer.config(com2)
```
