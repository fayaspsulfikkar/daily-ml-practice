```python
def greater_first(func):
    def wrap(a,b):
        if a<b:
            a,b = b, a

        return func(a,b)
    return wrap

@greater_first
def div(a,b):

    return a/b


def sub(a,b):

    return a-b

sub = greater_first(sub)

print(div(2,4))
print(sub(2,4))
```

```python
def log_deco(func):
    def wrap(*nums):
        print("Values: ",nums)
        result = func(*nums)
        return result
    return wrap

def greater_first(func):
    def wrap(a,b):
        if a<b:
            a,b = b, a

        return func(a,b)
    return wrap

@greater_first
@log_deco
def div(a,b):

    return a/b

@log_deco
def sub(a,b):

    return a-b

sub = greater_first(sub)

@log_deco
def add(*nums):
    return sum(nums)


print(div(2,4))
print(sub(2,4))
print(add(1,3,6))
```
