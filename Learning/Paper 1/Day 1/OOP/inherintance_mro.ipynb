```python
class A:

    def __init__(self):
        print("In A Init")

    def f1(self):
        print("F1 Works")


class B(A):

    def __init__(self):
        super().__init__()
        print("In B Init")

    def f2(self):
        super().f1()
        print("F2 Works")




son = B()


son.f2()
```
