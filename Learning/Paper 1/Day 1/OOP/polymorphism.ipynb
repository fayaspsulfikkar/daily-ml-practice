```python
class CreditCard:
    def pay(self):
        print("Processing Credit Card payment")


class UPI:
    def pay(self):
        print("Processing UPI payment")


class Cash:
    def pay(self):
        print("Processing Cash payment")


checkout1 = CreditCard()
checkout2 = UPI()
checkout3 = Cash()


checkout1.pay()
checkout2.pay()
checkout3.pay()

```

```python
def processpay(payment):
    return payment.pay()

processpay(CreditCard())
processpay(UPI())
processpay(Cash())
```
