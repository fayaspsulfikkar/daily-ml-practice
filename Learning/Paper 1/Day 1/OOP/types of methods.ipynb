```python
class Computer:

    brand = "BrotoType"

    def __init__(self, cpu, ram, ssd):
        self.cpu = cpu
        self.ram = ram 
        self.ssd = ssd

    def config(self):
        print("Config: ", self.cpu, self.ram, self.ssd)


    @classmethod
    def info(cls):
        return(cls.brand)



com1 = Computer("i7", "16GB","512SSD")
com2 = Computer("i5", "32GB", "1TB")


com1.config()
com2.config()
com2.info()

Computer.info()


```
