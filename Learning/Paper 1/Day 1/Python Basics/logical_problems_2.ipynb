```python
# Prime check: is_prime(n). Test 1, 2, 17, and 21 
```

```python
num = int(input("Enter a Number: "))

if num <= 1:
    print(f"{num} is Not Prime Number")
else:
    for i in range(2,num):
        if num % i == 0:
            print(f"{num} is not a Prime Number")
            break
    else:
        print(f"{num} is a Prime Number")
       

```

```python
# Unique numbers: from [1, 2, 2, 3, 4, 4, 5], get the unique values in order, then the values that appear exactly once 
```

```python
num = [1, 2, 2, 3, 4, 4, 5]
unique = []

for i in num:
    if i not in unique:
        unique.append(i)

print(sorted(unique))

```

```python
unique = []

for i in num:
    if num.count(i) == 1:
        unique.append(i)

print(unique)
```

```python
# Palindrome: a string ("madam") and a number (12321) 
```

```python
str = "madam"

if str[::-1] == str:
    print(f"{str} is a Palindrome")
else:
    print(f"{str} is not a Palindrome")
```

```python
number = 12321
original = number
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

if reverse == original:
    print(f"{original} is a Palindrome")
else:
    print(f"{original} is not a Palindrome")
```

```python
# Reverse a string: by slicing, by a loop, and with reversed 
```

```python
text = input("Enter a Text: ")
new_txt = ""

print(text[::-1])

for i in text:
     new_txt = i + new_txt
print(new_txt)

rever = ''.join(reversed(text))
print(rever)
```

```python
# Factorial: with a loop and with recursion. Test 5 
```

```python
num = int(input("Enter a Number: "))
factorial = 1

for i in range(1,num+1):
    factorial *= i

print(factorial)
```

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)


factorial(5)
```

```python
# Largest and second largest: without sort, on [4, 9, 2, 7, 5] 
```

```python
lis = [4, 9, 2, 7, 5]

print(max(lis))

larg = max(lis)
slarg = float('-inf')

for num in lis:
    if num != larg and num > slarg:
        slarg = num

print(slarg)
```
