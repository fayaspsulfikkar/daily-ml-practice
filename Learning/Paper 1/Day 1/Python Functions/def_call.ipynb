```python
def numbers():
    yield 1
    yield 2
    yield 3

result = numbers()

print(list(result))
```

```python
def calculate_total(*nums):
    total = 0

    for num in nums:
        total += num

    return total

calculate_total(10, 20, 30, 40, 50)
```

```python
def student_details(**kwargs):
    return kwargs

student_details(name="Fayas", age=24, course="AI/ML")
```

```python
def student_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student_details(name="Fayas", age=24, course="AI/ML")
```

```python
def example(name, age=20, *args, **kwargs):
    print(name)
    print(age)
    print(args)
    print(kwargs)

example("Fayas", 24, "Python", "ML", city="Calicut", country="India")
```

```python
def test(a, b=10):
    print(a, b)

test(5)
test(5, 20)
test(b=30, a=10)
```
