```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

```python
def calculate(a,b):
    addition = a + b
    subs = a - b
    multiplication = a * b 

    return addition, subs, multiplication

calculate(5,3)
```

```python
add, sub, mul = calculate(4,5)
print(add)
print(sub)
print(mul)
```

```python

def find_user(users, username):
    if username not in users:
        return None

    return username

usernames = ["Fayas","Velvet"]
status = find_user(usernames,"Ayush")

print(status)
```
