```python
def outer():
    message = "Hello from outer"

    def inner():
        print(message)

    inner()

outer()
```

```python
def outer():
    x = 10

    def inner():
        x = 20
        print(x)

    inner()
    print(x)

outer()
```

```python
count = 10

def update():
    global count
    count = 20

update()

print(count)
```

```python
numbers = [45, 42, 89, 23, 67, 10]

result = sorted(numbers, key=lambda x:x%10)

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x : x**2, numbers))

print(result)
```

```python
numbers = [1,2,3,4,5,6,7,8]

result = list(filter(lambda x: x%2==0, numbers))

print(result)
```

```python
numbers = [12, 45, 22, 67, 90, 33, 18]

result = list(filter(lambda x:x%3==0, numbers))

print(result)
```

```python
numbers = [2, 4, 6, 8, 10]

result = list(map(lambda x:(x*5)+10, numbers))

print(result)
```

```python
words = ["python", "AI", "machine", "data", "learning"]

result = sorted(words, key=lambda w:len(w))

print(result)
```
