```python
# 4. Sum of Digits
# - Take an integer as input.
# - Calculate the sum of all its digits.
# - Example: 1234 → 10
```

```python
num = int(input("Enter a Number: "))
original = num
total = 0 

while num>0:
    digit = num % 10 
    total += digit
    num = num // 10 

print("Sum is :", total)
```

```python
# Level 2 – Intermediate Logic
# 5. Reverse a Number
# - Take an integer as input.
# - Reverse its digits.
# - Example: 12345 → 54321
```

```python
num = int(input("Enter a Number: "))
original = num 
reverse = 0

while num>0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10 

print("Reversed Number: ",reverse)
```

```python
# 7. Check for Palindrome
# - Take a number as input.
# - Check whether it reads the same forward and backward.
# - Example: 121 → Palindrome
# - Example: 123 → Not a Palindrome
```

```python
num = int(input("Enter a Number: "))
original = num 
reverse = 0 

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Number Number is Palindrome")
else:
    print("Number is not Palindrome")
    
```

```python
# Level 3 – Logical Thinking
# 8. Find Factorial
# - Take a positive integer as input.
# - Calculate its factorial using a loop.
# - Example: 5 → 120
# - Remember: 5! = 5 × 4 × 3 × 2 × 1
```

```python
num = int(input("Enter a Number: "))
factorial = 1
original = num

while num > 0:
    factorial *= num
    num -= 1

print("Factorial: ", factorial)
```

```python
# 9. Check for Prime Number
# - Take an integer as input.
# - Determine whether it is a prime number.
# - Example: 7 → Prime
# - Example: 10 → Not Prime 
```

```python
num = int(input("Enter a Number: "))

if num ==2:
    print(num,"Number is Prime")
else:
    for i in range(2,num):
        if num % i == 0:
            print(num,"Number is not Prime")
            break
    else:
        print(num, "Number is prime")
```
