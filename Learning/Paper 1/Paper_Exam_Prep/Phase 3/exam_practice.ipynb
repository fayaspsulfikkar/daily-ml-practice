```python
import numpy as np

marks = np.array([
    [45, 60, 75],
    [80, 55, 90],
    [35, 70, 65],
    [95, 85, 80],
    [50, 45, 60]
])
```

```python
print(marks.min(axis=0))
print(marks.max(axis=1))
print(
    (marks - marks.min(axis=0)) /
    (marks.max(axis=0) - marks.min(axis=0))
)
new_array = print(
    (marks - marks.min(axis=0)) /
    (marks.max(axis=0) - marks.min(axis=0))
)

```
