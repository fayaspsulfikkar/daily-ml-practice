```python
A = np.array([
    [4, 2],
    [1, 3]
])
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)


print(eigenvalues)
print(eigenvectors)
```

```python
U, s, Vt = np.linalg.svd(A)

print(U)
print(s)
print(Vt)
```

```python
S = np.diag(s)

print(U @ S @ Vt)
```
