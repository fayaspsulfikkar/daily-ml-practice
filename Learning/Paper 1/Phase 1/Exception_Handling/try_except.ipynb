```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print(result)

except ValueError:
    print("Invalid input")

except ZeroDivisionError:
    print("Cannot divide by zero")

except BaseException as bs:
    print(bs)

except Exception:
    print("Something went wrong")
```

```python
try:
    with open("data.txt", "r") as file:
        content = file.read()
        print(content)

except FileNotFoundError as fnf:
    print(fnf)

except OSError:
    print("Something went wrong while reading the file")
```

```python
try:
    age = int(input("Enter Your Age: "))

    if age < 5:
        raise ValueError("Enter a valid age")

    print("Registration Successful")

except ValueError as e:
    print(e)
```

```python
class IsufficientBalance(Exception):
    pass

try:
    balance = 500
    withdrawal = 1000

    if withdrawal > balance:
        raise IsufficientBalance("Issuficient Balance")

except IsufficientBalance as e:
    print(e)    
    
finally:
    print("Program exited")
```
