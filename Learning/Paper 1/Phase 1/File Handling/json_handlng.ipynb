```python
import json

user = {
    "name": "Fayas",
    "age": 24,
    "skills": ["Python", "Machine Learning"],
    "is_student": True
}

with open("json_dump.json",'w') as file:
    json.dump(user, file, indent=4)
```

```python
with open("json_dump.json", 'r') as file:
    user = json.load(file)

print(user)
print(type(user))
```
