```python
import time


class Hello(Thread):

    def run(self):
        for i in range(5):
            time.sleep(0.5)
            print("Hello", i+1)


class Hi(Thread):

    def run(self):
        for i in range(5):
            time.sleep(0.5)
            print("Hi", i+1)


t1 = Hello()
t2 = Hi()

t1.start()
t2.start()
```

```python
from threading import Thread, Barrier

barrier = Barrier(2)


class Hello(Thread):
    def run(self):
        barrier.wait()
        for i in range(5):
            print("Hello", i + 1)


class Hi(Thread):
    def run(self):
        barrier.wait()
        for i in range(5):
            print("Hi", i + 1)


t1 = Hello()
t2 = Hi()

t1.start()
t2.start()

t1.join()
t2.join()
```

```python
from threading import Thread, Barrier
import time

barrier = Barrier(2)


class Hello(Thread):

    def run(self):
        barrier.wait()

        for i in range(5):
            print("Hello", i + 1)
            time.sleep(0.5)


class Hi(Thread):

    def run(self):
        barrier.wait()

        for i in range(5):
            print("Hi", i + 1)
            time.sleep(0.5)


t1 = Hello()
t2 = Hi()

t1.start()
t2.start()

t1.join()
t2.join()

print("Both threads completed!")
```

```python
from threading import Thread, Barrier, Lock
import time

barrier = Barrier(2)
lock = Lock()


class Hello(Thread):

    def run(self):
        barrier.wait()

        for i in range(5):
            with lock:
                print("Hello", i + 1)

            time.sleep(0.5)


class Hi(Thread):

    def run(self):
        barrier.wait()

        for i in range(5):
            with lock:
                print("Hi", i + 1)

            time.sleep(0.5)


t1 = Hello()
t2 = Hi()

t1.start()
t2.start()

t1.join()
t2.join()

print("Both threads completed!")
```
