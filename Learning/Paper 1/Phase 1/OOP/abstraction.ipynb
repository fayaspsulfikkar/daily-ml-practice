```python
from abc import ABC , abstractmethod

class A(ABC):

    @abstractmethod
    def show(self):
        pass
```

```python
class B(A):

    def show(self):
        pass
    
    def hi(self):
        print("Hi in B")

obj1= B()

obj1.hi()
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass

```

```python
class UPI(Payment):

    def pay(self):
        print("Payment through UPI")


class Card(Payment):

    def pay(self):
        print("Payment through Card")
```
