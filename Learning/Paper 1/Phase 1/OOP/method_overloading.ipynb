```python
class Account:

    def __init__(self, name, balance):
        self.name = name
        self.bal = balance

    def __str__(self):
        return f'{self.name} : {self.bal}'

    def __add__(self, other):
        return Account('Combined: ', self.bal + other.bal)

person1 = Account('Fayas', 500)
person2 = Account('Velvet', 2000)


print(person1)
print(person2)

combined = person1 + person2

print(combined)

```

```python
class Account:

    def __init__(self, name, balance):
        self.name = name
        self.bal = balance

    def __str__(self):
        return f'{self.name} : {self.bal}'

    def __gt__(self, other):
        return self.bal > other.bal

person1 = Account('Fayas', 500)
person2 = Account('Velvet', 2000)


print(person1)
print(person2)


if person1 > person2:
    print("Person 1 Pays")
else:
    print("Person 2 Pays")

```
