```python
class Computer:

    brand = "BrotoType"

    def __init__(self, cpu, ram, ssd):
        self.cpu = cpu
        self.ram = ram 
        self.ssd = ssd

    def config(self):
        print("Config: ", self.cpu, self.ram, self.ssd)


    @classmethod
    def info(cls):
        return(cls.brand)

    @staticmethod
    def gb_to_bytes(gb):
        return gb * (1024 ** 3)



com1 = Computer("i7", "16GB","512SSD")
com2 = Computer("i5", "32GB", "1TB")


com1.config()
com2.config()
com2.info()

Computer.info()


print(com2.gb_to_bytes(16))
```
