```python
for i in range(5):
    for j in range(i):
        print("* ",end=" ")
    print("")
```

```python
for i in range(4,0,-1):
    for j in range(i):
        print("* ",end=" ")
    print(" ")
```

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
for i in range(1,6):
    for j in range(0,i):
        print(i,end="")
    print()
```

```python
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()
```
