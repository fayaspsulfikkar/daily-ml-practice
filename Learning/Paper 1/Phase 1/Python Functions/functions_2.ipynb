```python
def calculate(a,b):
    addition = a + b
    subs = a - b
    multiplication = a * b 

    return addition, subs, multiplication

calculate(5,3)
```

```python
add, sub, mul = calculate(4,5)
print(add)
print(sub)
print(mul)
```

```python

def find_user(users, username):
    if username not in users:
        return None

    return username

usernames = ["Fayas","Velvet"]
status = find_user(usernames,"Ayush")

print(status)
```

```python
def square(num):
    return num ** 2

def cube(num):
    return num ** 3

def calculat(num, operator):
    return operator(num)


value = 5
result = calculat(value, square)

print(result)
```

```python
nums = [1,2,3,4,5,6,7,8]

def is_even(n):
    if n % 2 == 0:
        return n

even = filter(is_even, nums)

print(list(even))
```

```python
nums = [1,2,3,4,5,6,7,8]

is_even = lambda x: x % 2 == 0

even = filter(is_even, nums)

print(list(even))
```

```python
evens = filter(lambda x : x % 2 == 0, nums)
print(list(evens))
```

```python
words = ["apple", "hi", "banana", "cat", "elephant"]

result = filter(lambda x : len(x) > 4 , words)

print(list(result))
```

```python
evens = list(filter(lambda x: x % 2 == 0, nums))

doubles = map(lambda x: x * 2, evens)

print(list(doubles))
```

```python
from functools import reduce

evens = list(filter(lambda x: x % 2 == 0, nums))

doubles = map(lambda x: x * 2, evens)

su = reduce( lambda x,y : x + 7 , doubles )

print(su)
```
