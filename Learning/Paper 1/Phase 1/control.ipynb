```python
numbers = [1,2,3,4,5]

square = [x**2 for x in numbers]

print(square)
```

```python
even = [x for x in numbers if x%2==0]
print(even)
```

```python
even_sqr = [x**2 for x in numbers if x%2 == 0]
print(even_sqr)
```

```python
print(["Even" if x % 2 == 0 else "Odd" for x in numbers])
```

```python
words = ["a", "bb", "ccc"]

result = {word: len(word) for word in words}
print(result)
```

```python
words = ["a", "bb", "ccc", "dddd"]

result = {word: len(word) for word in words}
print(result)
```

```python
numbers = [1, 2, 3, 4, 5]

sqr = {num: num**2 for num in numbers}

print(sqr)
```

```python
numbers = [1, 2, 3, 4, 5]

result = {num: "Even" if num%2==0 else "Odd" for num in numbers}

print(result)
```

```python
students = [
    ("John", 85),
    ("Alex", 92),
    ("David", 78)
]
```

```python
students.sort(key=lambda student: student[1])
print(students)
```

```python
numbers = [1, 2, 3, 4]

result = map(lambda x: x ** 2, numbers)

print(list(result))
```
