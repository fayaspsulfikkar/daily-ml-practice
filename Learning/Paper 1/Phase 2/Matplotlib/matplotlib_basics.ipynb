```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2023,2024,2025,2026])
y = np.array([34,56,43,23])
y1 = np.array([78,67,56,69])
y3 = np.array([45,12,34,46])
```

```python
plt.plot(x,y,
         marker=".",
         ms=12,
         markerfacecolor = "DarkRed",
         markeredgecolor = "DarkRed",
         linestyle="dashdot",
         linewidth=2)

lin_style = dict(
         marker=".",
         ms=12,
         markerfacecolor="Darkred",
         markeredgecolor="Darkred",
         linestyle="dashdot",
         linewidth=2)
         

plt.plot(x,y1,**lin_style)

plt.plot(x,y3,**lin_style)

plt.title("Admission")
plt.xlabel("Years")
plt.ylabel("Students Admission Count")
plt.xticks(x)


plt.tight_layout()

plt.show()
```

```python
plt.plot(x,y,
         marker=".",
         ms=12,
         markerfacecolor = "DarkRed",
         markeredgecolor = "DarkRed",
         linestyle="dashdot",
         linewidth=2)

lin_style = dict(
         marker=".",
         ms=12,
         markerfacecolor="Darkred",
         markeredgecolor="Darkred",
         linestyle="dashdot",
         linewidth=2)

plt.grid(axis="both",
         linewidth=2,
         linestyle="dotted",
         alpha=0.8)
         

plt.plot(x,y1,**lin_style)

plt.plot(x,y3,**lin_style)

plt.title("Admission")
plt.xlabel("Years")
plt.ylabel("Students Admission Count")
plt.xticks(x)


plt.tight_layout()

plt.show()
```

```python
df = pd.read_csv("/media/raw/New Volume/BroCamp/Learning/Paper 1/Phase 2/Pandas/customers.csv")
df
```

```python
df["membership"].unique()
```

```python
df["membership"] = df["membership"].replace({"silver":"Silver",
                                                " Silver ":"Silver",
                                                "GOLD":"Gold"})
df["membership"].unique()
```

```python
membership_counts = df["membership"].value_counts()
```

```python
membership_counts.plot(kind="barh")

plt.xlabel("Number Of Members")
plt.ylabel("Membership")
plt.title("Membership Distribution")

plt.show()
```

```python
labels = membership_counts.index 
values = membership_counts.values
```

```python
plt.barh(labels, values,
         color="Darkred")

plt.yticks(labels)
plt.xticks(range(1,25,5))

plt.show()
```

```python
memberships = df["membership"].value_counts()
```

```python
memberships.plot(kind='pie',
                 autopct="%1.1f%%")

plt.show()
```

```python
label = memberships.index
value = memberships.values

plt.pie(value, labels=label,
        autopct="%1.1f%%",
        startangle=90
        )

plt.show()
```

```python

loyalty = df["loyalty_points"].dropna()

plt.figure(figsize=(9, 5))

plt.boxplot(
    loyalty,
    patch_artist=True,
    widths=0.35,
    boxprops=dict(facecolor="skyblue", edgecolor="black", linewidth=1.5),
    medianprops=dict(color="red", linewidth=2),
    whiskerprops=dict(color="black", linewidth=1.2),
    capprops=dict(color="black", linewidth=1.2),
    flierprops=dict(marker="o", markerfacecolor="red",
                    markersize=5, alpha=0.5)
)

plt.title("Distribution of Customer Loyalty Points",
          fontsize=15, fontweight="bold", pad=15)

plt.ylabel("Loyalty Points", fontsize=11)

plt.xticks([1], ["Customers"])

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
