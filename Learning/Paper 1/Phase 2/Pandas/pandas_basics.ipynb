```python
df = pd.read_csv("customers.csv")
df.sample(15)
```

```python
df["membership"].unique()
```

```python
df["membership"] = df["membership"].replace({"silver": "Silver",
                                             " Silver ": "Silver",
                                             "GOLD":"Gold"})

df.sample(20)
```

```python
df["membership"].unique()
```

```python
df["name"] = df["name"].str.lower()
df.sample(5)
```

```python
df["name"] = df["name"].str.upper()
df.sample(5)
```

```python
df["name"] = df["name"].str.capitalize()
df.sample(5)
```

```python
df["is_active"] = df["is_active"].astype(int)
df.sample(5)
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df = df.drop_duplicates()
df
```
