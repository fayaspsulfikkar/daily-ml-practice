```python
results = [2, 4, 6, 1, 3, 2, 5, 6, 4, 2,
           1, 3, 6, 5, 2, 4, 1, 6, 3, 2]
```

```python
poss_out = set(results)
print(poss_out)
```

```python
sample_space = len(results)
print(sample_space)
```

```python
for number in poss_out:
    fre = results.count(number)
    print(number, ":", fre)
```

```python
count2 = results.count(2)
print(count2)

prob_2 = count2 / len(results)

print(prob_2 * 100 )
```

```python
count_6 = results.count(6)

prob_6 = count_6 / len(results)

print(prob_6 * 100)
```

```python
odd_count = [x for x in results if x%2 != 0]

prob_odd = len(odd_count)/len(results)

print(prob_odd*100)
```

```python
gt_4 = [x for x in results if x > 4]

prob_gt4 = len(gt_4) / len(results)

print(prob_gt4 * 100)
```

```python
eodd_int_egt4 = set(odd_count).intersection(set(gt_4))
print(eodd_int_egt4)
```

```python
eodd_int_egt4 = len([x for x in odd_count if x in gt_4])

print(eodd_int_egt4)
```

```python
eodd_int_egt4 = len([x for x in odd_count if x in gt_4]) / len(results)

print(eodd_int_egt4)
```
