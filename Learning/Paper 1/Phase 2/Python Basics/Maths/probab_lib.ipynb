```python
import pandas as pd

data = {
    "Toss": range(1, 11),
    "Result": ["Heads", "Tails", "Heads", "Heads", "Tails",
               "Heads", "Tails", "Heads", "Tails", "Heads"]
}

df = pd.DataFrame(data)
df
```

```python
print(df["Result"].unique())
```

```python
print(df["Result"].value_counts())
```

```python
num_heads = (df["Result"]=="Heads").sum()

total_outcome = len(df)

prob_heads = num_heads / total_outcome
print(prob_heads*100)
```

```python
num_tails = (df["Result"]=="Tails").sum()

sample_space = len(df)

prob_tails = num_tails/sample_space

print(prob_tails*100)
```

```python
event_H = df[df["Result"]=="Heads"]
print(event_H)
```

```python
event_T = df[df["Result"]=="Tails"]
print(event_T)
```

```python
prob_even_h = len(event_H) / sample_space
print(prob_even_h*100)
```

```python
prob_event_t = len(event_T) / sample_space
print(prob_event_t*100)
```
