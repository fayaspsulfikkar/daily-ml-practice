```python
rng = np.random.default_rng(42)

roll = rng.integers(1,7, size=1000)

print(len(roll))
```

```python
event6 = [x for x in roll if x == 6]

prob_6 = len(event6) / len(roll)

print(prob_6*100)
```

```python
event_evn = [x for x  in roll if x%2 == 0]

prob_evn = len(event_evn) / len(roll)

print(prob_evn*100)
```

```python
event_gt4 = [x for x in roll if x > 4 ]

prob_gt4 = len(event_gt4)/ len(roll)

print(prob_gt4*100)
```

```python
# Q6. Sampling & Confidence Intervals
# Difficulty: 🟡 Beginner | 25 minutes

# Dataset:
# rng = np.random.default_rng(42)population = rng.normal(170, 10, 10000)


# Tasks:
# - [ ] Calculate the population mean and standard deviation.
# - [ ] Randomly select a sample of 100 observations.
# - [ ] Calculate the sample mean and standard deviation.
# - [ ] Repeat sampling 100 times.
# - [ ] Store each sample mean.
# - [ ] Visualize the sampling distribution.
# - [ ] Calculate a 95% confidence interval.
# - [ ] Compare the sample mean with the population mean.
# Expected output: Sampling distribution visualization and confidence interval.
```

```python
rng = np.random.default_rng(50)

population = rng.normal(170,10,1000)
```

```python
rand_sem = rng.choice(population, size=100, replace=False)
```

```python
s_mean = np.mean(rand_sem)
s_std = np.std(rand_sem)

print(s_mean)
print(s_std)
```

```python
sample_mean = []

for i in range(100):
    sample = rng.choice(population, size=100, replace=False)

    sample_mean.append(np.mean(sample))
```

```python
sns.kdeplot(sample_mean)

plt.show()
```
