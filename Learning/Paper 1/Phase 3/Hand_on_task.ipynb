```python
above60 = df[df["exam_score"]>60]

favourable = len(above60)

total = len(df["exam_score"].dropna())

probabiltity = favourable / total

print(probabiltity*100)
```

```python
below40 = df[df["exam_score"]<40]

favourable = len(below40)

total = len(df["exam_score"])

prob = favourable/total

print(prob*100)
```

```python
study5 = df[df["study_hours"]>5]

fav = len(study5)
total = len(df)

prob = fav/total

print(prob*100)
```

```python
# Calculate the probability that a student scores above 60 AND studies more than 5 hours.
```

```python
scor_60 = df[df["exam_score"]>60]
stud_5 = df[df["study_hours"]>5]

both = df[(df["exam_score"]>60) & (df["study_hours"]>5)]

favourable = len(both)

total = len(df)

prob = favourable/total

print(prob*100)
```

```python
# Calculate the probability that a student scores above 60 OR studies more than 5 hours.
```

```python
scab60 = df[df["exam_score"]>60]
stab5 = df[df["study_hours"]>5]

both = df[(df["exam_score"]>60) | (df["study_hours"]>5)]

favourable = len(both)

probabiltity = favourable/len(df)

print(probabiltity*100)
```

```python
import numpy as np

rng = np.random.default_rng(42)

normal = rng.normal(50, 10, 1000)
right_skewed = rng.exponential(10, 1000)
uniform = rng.uniform(0, 100, 1000)
```

```python
normal_ps = pd.Series(normal)
r_skewed = pd.Series(right_skewed)
uni_ps = pd.Series(uniform)
```
