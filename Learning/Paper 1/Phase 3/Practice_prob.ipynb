```python
# You roll a fair six-sided die once.

# Using Python:
# 1. Define the sample space.
# 2. Create an event containing all outcomes greater than 3.
# 3. Create another event containing all even outcomes.
# 4. Find the intersection of these two events.
# 5. Calculate the probability of their intersection.
# Expected output: The intersection and its probability.
```

```python
sample_space = [1,2,3,4,5,6]
event_gt3 = [x for x in sample_space if x>3]
event_ev = [x for x in sample_space if x%2==0]

inter = set(event_gt3) & set(event_ev)

prob_inter = len(inter) / len(sample_space)

print(prob_inter*100)
```

```python
# Q2 — Complementary Events
# Difficulty: Beginner | Time limit: 2 minutes

# A fair six-sided die is rolled once.

# Using Python:
# 1. Define an event containing outcomes greater than 4.
# 2. Find its complementary event (outcomes NOT greater than 4).
# 3. Calculate the probability of both events.
# 4. Verify that their probabilities add up to 1.
# Concepts tested: Complementary events, sample space, probability.
```

```python
die = [1,2,3,4,5,6]
even_gt4 = [x for x in die if x > 4]

prob_of_gt4 = len(even_gt4) / len(die)

com_gt4 = 1 - prob_of_gt4

print(com_gt4*100)
print(prob_of_gt4*100)

print(com_gt4+prob_of_gt4==1)
```

```python
# Q3 — Union of Events
# Difficulty: Beginner | Time limit: 3 minutes

# A fair six-sided die is rolled once.

# Using Python:
# 1. Create event A: outcomes greater than 3.
# 2. Create event B: odd outcomes.
# 3. Find the union of A and B.
# 4. Calculate P(A ∪ B).
# Concepts tested: Union, intersection, addition rule of probability.
```

```python
die = [1,2,3,4,5,6]

event_a = [x for x in die if x>3]
event_od = [x for x in die if x % 2 != 0]

union = set(event_a) | set(event_od)

prob_u = len(union) / len(die)
```

```python
# Q4 — Mutually Exclusive Events
# Difficulty: Beginner | Time limit: 3 minutes

# You roll a fair six-sided die once.

# Using Python:
# 1. Create event A: getting an even number.
# 2. Create event B: getting an odd number.
# 3. Find their intersection.
# 4. Determine whether they are mutually exclusive.
# 5. Verify the addition rule:
#    P(A ∪ B) = P(A) + P(B)
# Concepts tested: Mutually exclusive events, intersection, addition rule.
```

```python
die = [1,2,3,4,5,6]

event_e = [x for x in die if x % 2 == 0]
event_o = [x for x in die if x % 2 != 0]

inte = set(event_e) & set(event_o)


print(inte)
print(len(inte) == 0)
```
