```python
# Write plain-Python functions for mean, median, mode, range, variance and standard deviation. 
# Test odd and even list lengths, repeated values, and a list with an extreme value.
```

```python
def meen(x):
    meen = sum(x)/len(x)
    return meen

def mod(x):
    result = max(x, key=x.count)
    return result 


def mediian(x):
    n = len(x)
    x = sorted(x)

    if n % 2 == 1:
        
        median = x[n//2]
    else:
        median = (x[n//2-1] + x[n//2]) / 2

    return median

def rang(x):
    result = max(x) - min(x)
    return result

def var(x):
     mean = meen(x)

     dev = [(x-mean)**2 for x in x]
     result = sum(dev) / len(x)

     return result

def sdev(x):
    varr = var(x)

    stdev = varr ** 0.5

    return stdev
```

```python
data = [10, 15, 20, 15, 30]

print(meen(data))
print(mod(data))
print(mediian(data))
print(rang(data))
print(var(data))
print(sdev(data))
```
