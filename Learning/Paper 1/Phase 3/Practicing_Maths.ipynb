```python
# Write plain-Python functions for mean, median, mode, range, variance and standard deviation. 
# Test odd and even list lengths, repeated values, and a list with an extreme value.
```

```python
def meen(x):
    meen = sum(x)/len(x)
    return meen

def mod(x):
    result = max(x, key=x.count)
    return result 


def mediian(x):
    n = len(x)
    x = sorted(x)

    if n % 2 == 1:
        
        median = x[n//2]
    else:
        median = (x[n//2-1] + x[n//2]) / 2

    return median

def rang(x):
    result = max(x) - min(x)
    return result

def var(x):
     mean = meen(x)

     dev = [(x-mean)**2 for x in x]
     result = sum(dev) / len(x)

     return result

def sdev(x):
    varr = var(x)

    stdev = varr ** 0.5

    return stdev
```

```python
data = [10, 15, 20, 15, 30]

print(meen(data))
print(mod(data))
print(mediian(data))
print(rang(data))
print(var(data))
print(sdev(data))
```

```python
import numpy as np
import pandas as pd



arr = np.array([2,4,5,3,4,454,65,65,34,3212,43,5,643,12,32,4,5,4,34,658])

```

```python
p_var = np.var(arr)
s_var = np.var(arr, ddof=1)
p_dev = np.std(arr)
s_dev = np.std(arr, ddof=1)

print(p_var)
print(s_var)
print(p_dev)
print(s_dev)
```

```python
ps = pd.Series(arr)

p_var = ps.var(ddof=0)
s_var = ps.var()
p_dev = ps.std(ddof=0)
s_dev = ps.std()

print(p_var)
print(s_var)
print(p_dev)
print(s_dev)
```

```python
c1 = ["Head","Tails"]
c2 = ["Head","Tails"]

sample_space = [(x,y) for x in c1 for y in c2]
eventbh = [x for x in sample_space if x == ("Head","Head")]
eventah = [y for y in sample_space if "Head" in y]

prob_bh = len(eventbh) / len(sample_space)
prob_ah = len(eventah) / len(sample_space)

print(prob_bh*100)
print(prob_ah*100)
```
