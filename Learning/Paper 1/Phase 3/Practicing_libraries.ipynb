```python
# Task 1 — NumPy: Data Generation & Analysis
# Create a dataset representing the exam scores of 100 students.

# Requirements:
# 1. Generate 100 random exam scores using NumPy.
# 2. Scores should follow a normal distribution:
#    - Mean: 75
#    - Standard deviation: 12
# 3. Make sure all scores are between 0 and 100.
# 4. Calculate:
#    - Mean
#    - Median
#    - Standard deviation
#    - Minimum score
#    - Maximum score
# 5. Find how many students scored above 80.
# 6. Find how many students failed (score below 40).
# 7. Sort the scores in ascending order.
# Restrictions:
# - Use NumPy.
# - Don't use Pandas.
# - Don't use Python's built-in statistics module.
# What I'm evaluating:
# - NumPy array creation
# - Random number generation
# - Array manipulation
# - Boolean indexing
# - Statistical functions
```

```python
scores = np.random.normal(loc=75, scale=12, size=100)
scores = np.clip(scores,0,100)

print(scores)
```

```python
print(scores.mean())
print(np.median(scores))
print(scores.std())
print(scores.min())
print(scores.max())
```

```python
print(len(scores[scores > 80]))
```

```python
print(np.sum(scores < 40))
```

```python
print(np.sort(scores))
```

```python
data = {
    "customer_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "name": ["Alex", "John", "Sarah", "Mike", "Emma", "David", "Lisa", "Tom", "Anna", "Chris"],
    "age": [25, 32, np.nan, 28, 45, 32, 19, np.nan, 38, 29],
    "city": ["Calicut", "Kochi", "Calicut", "Bangalore", "Kochi", "Calicut", "Bangalore", "Kochi", "Calicut", "Bangalore"],
    "membership": ["Gold", "Silver", "Gold", "Bronze", "Silver", "Gold", "Bronze", "Gold", "Silver", "Bronze"],
    "loyalty_points": [850, 450, 920, 200, np.nan, 780, 150, 600, 500, 300],
    "purchase_amount": [5000, 2500, 7000, 1200, 3000, 4500, 800, 5500, 2800, 1500]
}

df = pd.DataFrame(data)
df.head(5)
```

```python
print("Rows: ",df.shape[0])
print("Columns: ",df.shape[1])
```

```python
print(len(df))
print(len(df.columns))
```

```python
df.isna().sum()
```

```python
age_median = df["age"].median()

df["age"] = df["age"].fillna(age_median)
df.isna().sum()
```

```python
lp_mean = df["loyalty_points"].mean()

df["loyalty_points"] = df["loyalty_points"].fillna(lp_mean)

df.isna().sum()
```

```python
df["membership"].unique()
```

```python
g_mem = df[df["membership"]=="Gold"]
print(g_mem)
```

```python
cgt500p = df[df["loyalty_points"]>500]
print(cgt500p)
```

```python
# Calculate the average purchase amount for each membership category.
```

```python
mem_cat = df.groupby("membership")["purchase_amount"].mean()
print(mem_cat)
```

```python
lp_cy = df.groupby("city")["loyalty_points"].mean()
lp_cy
```

```python
df.sort_values("purchase_amount",ascending=False)
```

```python
rng = np.random.default_rng(42)

data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],

    "sales": rng.integers(10000, 50000, 12),

    "customers": rng.integers(100, 500, 12),

    "profit": rng.integers(2000, 15000, 12),

    "satisfaction": rng.normal(7.5, 1, 12)
}

df = pd.DataFrame(data)

df
```

```python
month = df["month"]
sales = df["sales"]

plt.plot(month,sales,
         marker=".")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("sales")


plt.show()
```

```python
profit = df["profit"]

plt.barh(month,profit,)
plt.title("Monthly Sales")

plt.show()
```

```python
cus_count = df["customers"]
sales = df["sales"]

plt.scatter(cus_count,sales)

plt.title("Customer & Sales Correlation")
plt.xlabel("Customer Count")
plt.ylabel("Sales")

plt.show()
```

```python
cust_sat = df["satisfaction"]

plt.hist(cust_sat, bins=5,
         rwidth=0.95)

plt.show()
```

```python
month_sal = df["sales"]

plt.boxplot(month_sal)

plt.title("Monthly Sales BoxPlot")
plt.ylabel("Sales")


plt.show()
```

```python
# Create a 2×2 subplot layout:
# Position	Plot
# 1	Sales line plot
# 2	Profit bar chart
# 3	Customers scatter plot
# 4	Satisfaction histogram
```

```python
figure, axes = (plt.subplots(2,2))


axes[0,0].plot(month, sales,
               linewidth=1,
               linestyle="dotted",
               marker='o',
               markersize=3,
               label="Monthly Sales")
axes[0,0].grid(True, alpha=0.6)
axes[0,0].legend(fontsize=8)

axes[0,1].bar(month,profit)
axes[1,0].scatter(cus_count,sales)
axes[1,1].hist(cust_sat,bins=5,rwidth=0.95)

axes[0,0].tick_params(axis='x', rotation=65)
axes[0,1].tick_params(axis='x', rotation=65)


plt.tight_layout()

plt.show()
```

```python
mean_sales = np.mean(sales)


plt.figure(figsize=(8,10))
plt.hist(sales, bins=5, edgecolor="black")

plt.axvline(mean_sales, color="red", linestyle="dashed",
            linewidth=2, label="Mean Sales")

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```
