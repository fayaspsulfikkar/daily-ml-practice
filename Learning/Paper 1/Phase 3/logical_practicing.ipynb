```python
# Q1. Prime Number Checker 🟢

# Write a program that checks whether a given number is prime or not.

# Example:
# Input: 17Output: PrimeInput: 20Output: Not Prime
```

```python
num = int(input("Enter a num: "))

if num >=1:
    for i in range(2, int(num**0.5)+1):
        if num%i == 0:
            print(f'{num} is not a prime Number')
            break
    else:
        print(f'{num} is a prime number')
else:
    print(f'{num} is not a prime number')
```

```python
# Q2. Find Unique Numbers 🟢

# Given a list, return only the numbers that appear exactly once.
# Input: [1, 2, 2, 3, 4, 4, 5]Output: [1, 3, 5]
```

```python
x = [1, 2, 2, 3, 4, 4, 5]

unique = []

for i in x:
    if x.count(i) == 1:
        unique.append(i)

print(unique)
```

```python
# Q3. Find the Second Largest Number 🟢

# Find the second-largest distinct number without using sort() or sorted().
# Input: [10, 5, 20, 8, 20, 15]Output: 15
```

```python
x = [10, 5, 20, 8, 20, 15]

large = float('-inf')
second = float('-inf')


for num in x:
    if num > large:
        second = large
        large = num
    elif num < large and num > second:
        second = num

print(second)
```

```python
# Palindrome Checker 🟢

# Check whether a given string is a palindrome.

# A palindrome reads the same forward and backward.
# Input: "madam"Output: TrueInput: "python"Output: False
```

```python
x = input("Enter a string: ")

if x[::-1] == x:
    print(x[::-1]==x)
else:
    print(x[::-1]==x)
```

```python
# Q5. Count Vowels and Consonants 🟢

# Given a string, count the number of vowels and consonants.
# Input: "Artificial Intelligence"Output:Vowels: 10Consonants: 12
```

```python
text = input("Enteer a Text: ")

text_l = text.lower()
v_count = 0 
c_count = 0

for i in range(len(text_l)):
    if text_l[i] in "aeiou":
        v_count += 1
    elif text_l[i].isalpha():
        c_count += 1

print(v_count)
print(c_count)
```
