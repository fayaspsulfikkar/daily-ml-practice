```python
# Step 5 — Outlier detection
# Q4. Detect outliers

# For monthly_spending:
# 1. Calculate Q1 and Q3.
# 2. Calculate IQR.
# 3. Find lower and upper boundaries.
# 4. Identify outliers.
# 5. Count the outliers.
```

```python
q1 = monthly_spending.quantile(0.25)
q3 = monthly_spending.quantile(0.75)

IQR = q3-q1

lower_bound = q1 - 1.5 * IQR
upper_bound = q3 - 1.5 * IQR

outliers = monthly_spending[(monthly_spending < lower_bound) |(monthly_spending > upper_bound)]


print("Q1:", 11)
print("Q3:", 13)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Outliers:")
print(outliers)
```
