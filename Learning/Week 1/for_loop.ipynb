```python
#Count even and odd numbers
numbers = [12, 101, 7, 89, 23, 56, 91, 34]
even_count=0
odd_count=0
for i in numbers:
    if i%2==0:
        even_count += 1
    else:
        odd_count += 1
print(even_count,odd_count)
```

```python
#reverse_the_list
numbers = [1, 2, 3, 2, 4, 5, 3, 6, 1]
rev=[]
for i in range(len(numbers)-1,-1,-1):
    rev.append(numbers[i])
print(rev)
```

```python
#Find duplicates
numbers = [1, 2, 3, 2, 4, 5, 3, 6, 1]
seen=[]
duplicates=[]

for i in numbers:
    if i in seen:
        if i not in duplicates:
            duplicates.append(i)
    else:
        seen.append(i)
print(duplicates)
```

```python
#find the second largest unique numbers
numbers = [89, 45, 23, 67, 89, 34, 89, 12]
largest=numbers[0]
second=None 

for i in numbers:
    if largest<i:
        largest=i

for i in numbers:
    if i==largest:
        continue
    if second is None or i > second:
        second=i
print(second)
```

```python
#character_count
text = "hello"
char_counts = {}

for char in text:
    if char in char_counts:
        char_counts[char] += 1
    else:
        char_counts[char] = 1

print(char_counts)
```

```python
#printing_Unique_numbers
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 5]
uniq_num=[]

for num in numbers:
    if num not in uniq_num:
        uniq_num.append(num)
print(uniq_num)
```

```python
#remove_duplicates
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 5]
no_duplicates=[]

for num in numbers:
    if num not in no_duplicates:
        no_duplicates.append(num)
print(no_duplicates)
```

```python
#print pairs
numbers = [2, 4, 6, 8, 10]
target_sum = 12

for i in range(len(numbers)):
    for j in range(i+1,len(numbers)):
        if numbers[i]+numbers[j]==target_sum:
            print(f"({numbers[i]},{numbers[j]})")
```
