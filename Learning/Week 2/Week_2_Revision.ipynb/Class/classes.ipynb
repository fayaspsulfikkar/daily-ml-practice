```python
#Build Book (title, author, pages) with a describe() method and a class attribute that counts books created. 
# Hints: Where does self live? What changes if the counter is set inside __init__?
```

```python
class BuildBook:
    book_count = 0

    def __init__(self,title,author,pages):
        self.title = title
        self.author = author
        self.pages = pages
        BuildBook.book_count +=1
    def decribe(self):
        print("Title: ",self.title)
        print("Author: ",self.author)
        print("Pages: ",self.pages)
        print()



book1 = BuildBook("Harry Potter","J-K Rownling",650)
book2 = BuildBook("Ayush Potter","J-K Hussain",1250)

book1.decribe()
book2.decribe()
print(BuildBook.book_count)

```

```python
def even_numbers_generator():
    # Iterate from 1 to 10 inclusive
    for num in range(1, 11):
        if num % 2 == 0:
            yield num

# Create the generator object
evens = even_numbers_generator()

# Print the numbers by iterating over the generator
print(type(evens))
print("This is even number : ", evens)
```
