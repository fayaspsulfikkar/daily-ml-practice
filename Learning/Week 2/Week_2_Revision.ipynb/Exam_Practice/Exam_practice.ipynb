```python
#Create 3 Student objects and print the one with the highest marks.

class Students:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark


fayas = Students("Fayas", 60)
velvet = Students("Velvet", 90)
mahir = Students("Mahir", 99)

students = [fayas, velvet, mahir]

highest = max(students, key=lambda student: student.mark)

print(highest.name, highest.mark)
```

```python
#Add __str__ so print(student) shows a nice sentence.
class Students:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

    def __str__(self):
        return f"Student {self.name} Have {self.mark} Marks"

student1 = Students("Fayas",90)
student2 = Students("Velvet",99)

print(student2)
```

```python
#Make a class with a destructor that prints a message when the object is deleted.

class Dest:
    def __init__(self,name):
        self.name = name
    def __del__(self):
        print("Print Student deleted")


a = Dest("Fayas")
b = Dest("Velvet")
c = Dest("Vel")

del c
del b
del a

```

```python
#Use *args to write a function that sums any number of values.

def add(*num):
    total = 0

    for values in num:
        total += values

    return total

print(add(2,3,4,5))
```

```python
#Write a function that returns both the smallest and largest number of a list.


def smallLarge(numbers):
    smallest = min(numbers)
    largest = max(numbers)

    return smallest,largest

numbers=[34,54,56,12,34,65,67,88,90,45,51,37]

print(smallLarge(numbers))
```
