```python
# Exercise 1: choose a message with if, elif, and else
score = 78
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 50:
    grade = "C"
else:
    grade = "Needs improvement"
print("Grade:", grade)
```

```python
# Exercise 2: use nested conditions for an admission rule
age = 20
has_id = True
if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("Bring a valid ID")
else:
    print("Entry not allowed")
```

```python
# Exercise 3: use a for loop to calculate a total
prices = [12, 35, 60, 18]
total_price = 0
for price in prices:
    total_price += price
print("Total:", total_price)
```

```python
# Exercise 4: use a while loop with a clear stopping condition
number = 1
squares = []
while number <= 5:
    squares.append(number ** 2)
    number += 1
print(squares)
```

```python
# Exercise 5: use continue and break to control a loop
valid_scores = []
for score in [88, -1, 92, 101, 75]:
    if score == -1:
        continue
    if score > 100:
        break
    valid_scores.append(score)
print(valid_scores)
```

```python
# Practice challenge: find the first number divisible by both 3 and 7
for number in range(1, 101):
    if number % 3 == 0 and number % 7 == 0:
        print("First match:", number)
        break
```
