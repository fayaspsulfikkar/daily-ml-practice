```python
# Exercise 4: read and update a dictionary safely
student = {"name": "Meera", "score": 95, "city": "Thrissur"}
student["score"] = 98
print(student["name"])
print(student.get("email", "Email not available"))
```

```python
# Exercise 5: count word frequency with a dictionary
words = ["python", "code", "python", "data", "code", "python"]
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
print(counts)
```

```python
# Practice challenge: group names by their first letter
names = ["Asha", "Arun", "Meera", "Maya", "Ravi"]
groups = {}
for name in names:
    first_letter = name[0]
    groups.setdefault(first_letter, []).append(name)
print(groups)
```
