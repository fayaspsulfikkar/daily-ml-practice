```python
# Exercise 1: create a function that returns a value
def rectangle_area(width, height):
    return width * height


print(rectangle_area(5, 3))
```

```python
# Exercise 2: use default and keyword parameters
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"


print(greet("Asha"))
print(greet(name="Ravi", greeting="Good morning"))
```

```python
# Exercise 3: accept a variable number of values with *args
def average(*numbers):
    return sum(numbers) / len(numbers) if numbers else 0


print(average(70, 80, 90))
```

```python
# Exercise 4: see the difference between local and global scope
course_name = "Python"

def show_course():
    lesson_name = "Functions"
    print(course_name, "-", lesson_name)


show_course()
```

```python
# Exercise 5: build reusable validation logic
def is_valid_score(score):
    return isinstance(score, (int, float)) and 0 <= score <= 100


for score in [88, -5, "90", 101]:
    print(score, "=>", is_valid_score(score))
```

```python
# Practice challenge: return the highest-scoring student
def top_student(records):
    return max(records, key=lambda record: record["score"])


students = [{"name": "Asha", "score": 88}, {"name": "Meera", "score": 95}]
print(top_student(students))
```
