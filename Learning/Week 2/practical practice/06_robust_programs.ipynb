```python
# Exercise 1: catch an invalid integer conversion
value = "twelve"
try:
    number = int(value)
except ValueError:
    print(f"Cannot convert {value!r} to an integer")
```

```python
# Exercise 2: handle more than one expected error
numerator = "12"
denominator = "0"
try:
    result = int(numerator) / int(denominator)
except ValueError:
    print("Use whole-number text")
except ZeroDivisionError:
    print("A denominator cannot be zero")
```

```python
# Exercise 3: use else and finally for clear cleanup flow
values = [10, 20, 30]
try:
    print(values[1])
except IndexError:
    print("Index not available")
else:
    print("Read completed successfully")
finally:
    print("This message always runs")
```

```python
# Exercise 4: define a custom exception for a business rule
class InvalidScoreError(Exception):
    pass


def save_score(score):
    if not 0 <= score <= 100:
        raise InvalidScoreError("Score must be between 0 and 100")
    return f"Saved score: {score}"


try:
    print(save_score(105))
except InvalidScoreError as error:
    print(error)
```

```python
# Exercise 5: return a safe result instead of stopping a program
def safe_divide(first, second):
    try:
        return first / second
    except ZeroDivisionError:
        return None


print(safe_divide(10, 2))
print(safe_divide(10, 0))
```
