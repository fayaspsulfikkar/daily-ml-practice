```python
# Exercise 1: create a class with instance properties and a constructor
class Student:
    school_name = "BroCamp"

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def summary(self):
        return f"{self.name}: {self.score}"


asha = Student("Asha", 88)
print(asha.summary())
print(Student.school_name)
```

```python
# Exercise 2: use inheritance to reuse common behavior
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I am {self.name}."


class Mentor(Person):
    def review(self, project):
        return f"{self.name} reviewed {project}."


mentor = Mentor("Ravi")
print(mentor.introduce())
print(mentor.review("Sales report"))
```

```python
# Exercise 3: override a method for polymorphic behavior
class Notification:
    def send(self, message):
        return f"Notification: {message}"


class EmailNotification(Notification):
    def send(self, message):
        return f"Email sent: {message}"


for channel in [Notification(), EmailNotification()]:
    print(channel.send("Practice completed"))
```

```python
# Exercise 4: use a property to protect a value
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero")
        self._celsius = value


room = Temperature(25)
print(room.celsius)
```

```python
# Exercise 5: destructor demonstration (do not rely on it for important cleanup)
class PracticeSession:
    def __init__(self, topic):
        self.topic = topic
        print("Started:", topic)

    def __del__(self):
        print("Session object cleaned up:", self.topic)


session = PracticeSession("OOP")
del session
```
