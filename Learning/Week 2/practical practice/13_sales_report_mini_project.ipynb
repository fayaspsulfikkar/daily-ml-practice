```python
# Project: build a small sales report from a CSV file
from pathlib import Path
import csv

current_folder = Path.cwd().resolve()
project_root = next(folder for folder in [current_folder, *current_folder.parents] if (folder / "data").is_dir())
sales_path = project_root / "data" / "sales.csv"
print("Source file:", sales_path.name)
```

```python
# Step 1: load and convert the raw CSV data
with sales_path.open(newline="", encoding="utf-8") as file:
    raw_sales = list(csv.DictReader(file))

sales = [
    {
        **row,
        "quantity": int(row["quantity"]),
        "unit_price": float(row["unit_price"]),
    }
    for row in raw_sales
]
print(sales[0])
```

```python
# Step 2: calculate the total revenue for every sale
for sale in sales:
    sale["revenue"] = sale["quantity"] * sale["unit_price"]
print(sales[0])
```

```python
# Step 3: summarize revenue by category
category_revenue = {}
for sale in sales:
    category = sale["category"]
    category_revenue[category] = category_revenue.get(category, 0) + sale["revenue"]
print(category_revenue)
```

```python
# Step 4: find the highest-revenue transaction
top_sale = max(sales, key=lambda sale: sale["revenue"])
print("Top sale:", top_sale)
```

```python
# Step 5: write a reusable text report
report_lines = ["SALES REPORT", "=" * 20]
report_lines.append(f"Total revenue: {sum(sale['revenue'] for sale in sales):.2f}")
for category, revenue in sorted(category_revenue.items()):
    report_lines.append(f"{category.title()}: {revenue:.2f}")

report_path = project_root / "data" / "sales_report.txt"
report_path.write_text("\n".join(report_lines) + "\n", encoding="utf-8")
print(report_path.read_text(encoding="utf-8"))
```
