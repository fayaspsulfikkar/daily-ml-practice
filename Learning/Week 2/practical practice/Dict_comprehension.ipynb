```python
#Make a dict where each number from 1 to 5 is a key and its square is the value.

square = {num:num**2 for num in range(1,6)}
print(square)
```

```python
#Name and first letter
#Given names = ["Asha", "Ravi", "Meera"], make a dict of each name and its first letter.

names = ["Asha", "Ravi", "Meera"]

name_dict = {name:name[0] for name in names}
print(name_dict)
```

```python
#Given nums = [1, 2, 3, 4], make a dict that says "even" or "odd" for each number.

nums = [1, 2, 3, 4]

dict_nums = {num: "even" if num%2==0 else "odd" for num in nums }

print(dict_nums)
```

```python
#Given prices = {"pen": 10, "book": 50, "bag": 200}, keep only items that cost more than 20.

prices = {"pen": 10, "book": 50, "bag": 200}

price = {key:pri for key,pri in prices.items() if pri>20 }
print(price)
```

```python
#Given prices = {"pen": 10, "book": 50, "bag": 200}, make a new dict with every price doubled.

prices = {"pen": 10, "book": 50, "bag": 200}

double_prices={key:value*2 for key,value in prices.items() }

print(double_prices)
```
