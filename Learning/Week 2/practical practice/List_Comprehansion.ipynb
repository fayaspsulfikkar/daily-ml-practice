```python
#Make a list of the squares of the numbers 1 to 10.

square = [i*i for i in range(1,11)]

print(square)

```

```python
#Given words = ["apple", "fig", "banana", "kiwi", "cherry"], keep only words longer than 4 letters and make them uppercase.

words = ["apple", "fig", "banana", "kiwi", "cherry"]

new_words=[i.upper() for i in words if len(i)>4]

print(new_words)
```

```python
#Given scores = [88, -1, 92, None, 75, 101, 60], keep only valid scores (numbers from 0 to 100).

scores = [88, -1, 92, None, 75, 101, 60]

valid_scores = [i for i in scores if i is not None and  i>=0 and i<=100]

print(valid_scores)
```

```python
#Given temps = [22, 35, 18, 41, 30], make a list that says "hot" if the temperature is above 30, otherwise "ok"

temps = [22, 35, 18, 41, 30]

list = ["hot" if temp>30 else "ok" for temp in temps]

print(list)
```
