```python
#Make a lambda that doubles a number. Test it with 7

dub = lambda num : num*2

dub(8)
```

```python
#Make a lambda that adds two numbers.

add = lambda num,num2: num+num2

add(8,4)
```

```python
#Make a lambda that returns "Even" or "Odd" for a number.

odd_even = lambda num : "even" if num%2==0 else "Odd"

odd_even(9)
```

```python
#Make a lambda that takes a name and returns "Hello, <name>!"

greet = lambda name : print("Hello", name)

greet("Fayas")
```
