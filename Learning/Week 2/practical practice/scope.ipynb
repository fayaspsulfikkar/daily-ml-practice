```python
def check_scope():
    def do_local():
        test="local test"
    def do_non_local():
        nonlocal test
        test="do_non_local test"
    def do_global():
        global test
        test="do_global test"

    test="default"
    do_local()
    print("test value after do_local: ",test)

    do_non_local()
    print("test value after do_non_local: ",test)

    do_global()
    print("test value after do_global: ",test)

check_scope()
print("test value after main",test)
```

```python
#To modify a global variable inside a function, use global:

count = 0

def increase():
    global count
    count += 1

increase()

print(count)
```
