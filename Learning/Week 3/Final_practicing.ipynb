Axis-Based Summaries

```python
marks = np.array([
    [80, 70, 90],
    [60, 85, 75],
    [90, 95, 85]
])
```

```python
# Each row represents a student. Each column represents a subject.
# Predict the results:
# Operation	Meaning
# marks.sum()	Sum of all marks
# marks.sum(axis=0)	Total marks for each subject
# marks.sum(axis=1)	Total marks for each student
# marks.mean(axis=0)	Average mark for each subject
# marks.max(axis=1)	Highest mark for each student
```

```python
print(marks.sum())
print(marks.sum(axis=0))
print(marks.sum(axis=1))
print(marks.mean(axis=0))
print(marks.max(axis=1))
```

```python
#Compare the number of customers in different categories using a suitable categorical column.

customer_df = pd.read_csv("customers.csv")

customer_df["membership"] = customer_df["membership"].replace("GOLD","Gold")
customer_df["membership"] = customer_df["membership"].replace(" Silver ","Gold")
customer_df["membership"] = customer_df["membership"].replace("silver","Silver")

categories = customer_df["membership"].value_counts()

plt.bar(categories.index, categories.values)

plt.show()
```
