```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,51)


figures, axes = plt.subplots(2,2)

axes[0,0].plot(x,x**2, color="red")
axes[0,0].set_title("x**2")

axes[0,1].plot(x,x**3)
axes[0,1].set_title("x**3")

axes[1,0].plot(x,x**4)
axes[1,0].set_title("x**4")

axes[1,1].plot(x,x**5)
axes[1,1].set_title("x**5")
#axes[1,1].set_xlim(10)

plt.tight_layout()
plt.show()
```
