```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
hours_studied = np.array([1,3,4,2,5,6,4,8,3,9])
scores = np.array([15,30,40,25,55,62,49,87,40,92])

hours_studied1 = np.array([4,3,6,7,8,9,6,2,1,5])
scores1 = np.array([45,30,60,75,85,92,29,7,10,52])

plt.scatter(hours_studied,scores,
            color="Darkred",
            s=50,
            label="Class A"
            )

plt.scatter(hours_studied1,scores1,
            color="Darkgreen",
            s=50,
            label="Class B"
            )

plt.title("Students Mark Correletion",
          size=20,
          family="serif",
          fontstyle="italic",
          )
plt.xticks(color="darkblue")
plt.yticks(color="darkblue")

plt.legend()
plt.show()
```
