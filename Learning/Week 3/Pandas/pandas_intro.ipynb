```python
data = ["A",False,300.3,400.4]
series = pd.Series(data,index = ["A", "B", "C", "D"])

print(series)
```

```python

print(series.loc["D"])
series.loc["D"] = 500
print(series)
```

```python
#integer_locator

data = ["A",False,300.3,400.4]
series = pd.Series(data,index = ["A", "B", "C", "D"])

print(series)

series.iloc[0] = 100

print(series)
```

```python
data1 = [100,200,300,400,500]
series = pd.Series(data1)
print(series[series >= 250])
```

```python
calories = {
    "Day1": 1750,
    "Day2": 2100,
    "Day3": 1800
}

print(calories["Day1"])
```

```python
series = pd.Series(calories)
print(series)
```

```python

print(series.loc["Day1"])
```

```python
series.loc["Day3"]+=500
print(series)
```

```python
print(series[series>2000])
```
