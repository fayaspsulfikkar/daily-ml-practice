```python
# ### Q4 · Pandas — New column and groupby (14 min)

# - **Goal:** build a label column and summarise by group.
# - **Data:** build a DataFrame `emp` with 6 rows:
#     - `name`: Asha, Ben, Chen, Dia, Eli, Fay
#     - `dept`: Sales, IT, IT, Sales, HR, IT
#     - `salary`: 40000, 55000, 45000, 45000, 38000, 60000
# - **Steps:**
#     - (a) Add a column `band` using `np.where`: "high" if salary is 50000 or more, otherwise "low".
#     - (b) Count how many rows are high and how many are low.
#     - (c) Calculate the average salary per `dept`, sorted from highest to lowest.
#     - (d) Count the employees in each `dept`.
#     - (e) Show the IT employees who earn more than 50000.
# - **Check yourself:** (b) low 4, high 2. (c) IT 53333.33, Sales 42500.0, HR 38000.0. (d) IT 3, Sales 2, HR 1. (e) Ben and Fay.
```

```python
# ### Q6 · Matplotlib — Legend and subplots (13 min)

# - **Goal:** put two lines on one chart, then split them across two charts.
# - **Data:** `x = np.arange(1, 6)`, `linear = 2 * x`, `square = x ** 2`.
# - **Steps:**
#     - (a) Figure 1: plot both lines on the same axes with different markers. Label each line, add a legend, a title and both axis labels.
#     - (b) Figure 2: use `plt.subplots(1, 2, figsize=(10, 4))`. Left: line plot of `linear`. Right: bar chart of `square`. Give each its own title.
#     - (c) Call `plt.tight_layout()` before showing.
# - **Check yourself:** `linear` is `[ 2  4  6  8 10]`, `square` is `[ 1  4  9 16 25]`. The two lines meet at x = 2 (both are 4). The legend has 2 entries. Figure 2 has two charts side by side.
```

```python
x = np.arange(1,6)
linear = 2 * x
square = x **2

plt.plot(x, linear, marker="o", label="Linear")
plt.plot(x, square, marker="s", label="Square")
plt.title("Linear vs Square")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

fig, axes = plt.subplots(1,2, figsize=(10,4))

axes[0].plot(x,linear)
axes[0].set_title("Linear")

axes[1].bar(x,square)
axes[1].set_title("Square")

plt.tight_layout()

plt.show()
```

```python
# ### Q7 · Mixed (NumPy + Pandas + Matplotlib) — Label, group, chart (13 min)

# - **Goal:** create a pass/fail label with NumPy, summarise by group, and chart the result.
# - **Data:** `scores = np.array([34, 78, 56, 91, 45, 67, 82, 59, 73, 88])` and 
#             `subject = ["Math", "Sci", "Math", "Sci", "Math", "Sci", "Math", "Sci", "Math", "Sci"]`.
# - **Steps:**
#     - (a) Build a DataFrame with columns `subject` and `score`.
#     - (b) Add a column `result` using `np.where`: "pass" if score is 60 or more, otherwise "fail".
#     - (c) Count the pass and fail rows for each subject.
#     - (d) Calculate the average score for each subject.
#     - (e) Draw a bar chart of the averages with a title and both axis labels.
# - **Check yourself:** Math: 2 pass, 3 fail, average 58.0. Sci: 4 pass, 1 fail, average 76.6.
```

```python
scores = np.array([34, 78, 56, 91, 45, 67, 82, 59, 73, 88])
subject = ["Math", "Sci", "Math", "Sci", "Math", "Sci", "Math", "Sci", "Math", "Sci"]

df = pd.DataFrame({
    "scores":scores,
    "subject":subject
})
df["result"] =  np.where(scores>60,"pass","fail")
count = df.groupby(["subject","result"]).size()
print(count)

avg = df.groupby("subject")["scores"].mean()
print(avg)

plt.bar(avg.index,avg.values)
plt.title("Subject's Mark")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()
```

```python
# ### Q8 · Mixed (NumPy + Pandas + Matplotlib) — Random heights (14 min)

# - **Goal:** generate repeatable random data, check it two ways, and plot it.
# - **Data:** `rng = np.random.default_rng(7)` and `heights = rng.integers(150, 190, size=30)`.
# - **Steps:**
#     - (a) Print the shape, minimum and maximum of `heights`.
#     - (b) Build a DataFrame with `id` (1 to 30) and `height`.
#     - (c) Add a True/False column `tall` for heights of 170 or more.
#     - (d) Count the tall people with Pandas, then again with NumPy. The two numbers must match.
#     - (e) Draw a histogram with bin edges `[150, 160, 170, 180, 190]`, a red vertical line at the mean (`plt.axvline`), a legend, a title and both axis labels.
# - **Check yourself:** shape `(30,)`, minimum 150, maximum 189. Mean about 170.83. Tall count is 15 both ways. Bar counts `[ 5. 10.  5. 10.]`.
```

```python
rng = np.random.default_rng(7)
heights = rng.integers(150,190,size=30)

print(heights.shape, heights.min(), heights.max())

df = pd.DataFrame({
    "id": np.arange(1,31),
    "Height": heights
})


df["Tall"] = df["Height"]>=170

print(df["Tall"].sum(), (heights >= 170).sum())

plt.hist(heights,bins=[150,160,170,180,190],
         edgecolor = "black",
         rwidth = 0.9)
plt.axvline(heights.mean(),color="red",label="Mean")
plt.title("Heights Hist")
plt.xlabel("Heights")
plt.ylabel("Counts")
plt.legend()

plt.show()
```
