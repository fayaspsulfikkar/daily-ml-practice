```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
waiting_time = np.random.exponential(scale=1/5.,size=1000)
print(waiting_time)
avg_wait = sum(waiting_time)/len(waiting_time)
avg_wait
```

```python
avg_wait_sec = avg_wait*60
avg_wait_sec
```

```python
plt.hist(waiting_time,
         bins=30,
         rwidth=0.90)

plt.show()
```
