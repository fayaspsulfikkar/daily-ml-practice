```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("calls_data.csv")
df
```

```python
discrete = df["number_of_calls"]
continous = df["waiting_time_seconds"]

print("Dicrete Variable: \n",discrete)
print("Continous Variable: \n",continous)
```

```python
total_p = df.shape[0]
num_5_calls =(df["number_of_calls"]==5).sum()
print(num_5_calls)

P_5 = num_5_calls/total_p
print("P(5) Calls: ", P_5)
```

```python
df.shape[0]
```

```python
total_po = df.shape[0]
num_5_or_less_calls = (df["number_of_calls"]<=5).sum()


p_5rl = num_5_or_less_calls/total_po

print("P(X<=5): ",p_5rl)
```

```python
waiting_times = df["waiting_time_seconds"]


plt.hist(waiting_times,
         bins=6,
         rwidth=0.9,
         density=True)
plt.title("Waiting time distribution")
plt.xlabel("Waiting Time (seconds)")
plt.ylabel("Frequency")

plt.show()
```
