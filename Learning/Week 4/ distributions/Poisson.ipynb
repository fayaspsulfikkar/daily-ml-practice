```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
calls = np.random.poisson(lam=5,size=1000)
calls
```

```python
mean_calls = sum(calls)/len(calls)
print(mean_calls)
```

```python
plt.hist(calls,
         bins=range(0,11),
         rwidth=0.90,
         color="black")

plt.xlabel("Number of calls per minute")
plt.ylabel("Frequency")
plt.title("Poisson Distribution: Calls per Minute")

plt.show()
```
