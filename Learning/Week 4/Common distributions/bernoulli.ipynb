```python
# Suppose the student takes the same 20-question test 1,000 times.

# Write Python code to:
# 1. Simulate 1,000 tests, each containing 20 questions.
# 2. Store the results in test_results.
# 3. Calculate the average number of correct answers across all 1,000 tests.
# 4. Print the average.
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
test_results = np.random.binomial(20,0.7,size=1000)
avg_results = sum(test_results)/len(test_results)
print("The Average is: ",avg_results)
```

```python
# Using your existing test_results variable, draw a histogram to show how frequently the student gets different numbers of questions correct.
```

```python
plt.hist(test_results,
         bins=13,
         rwidth=0.90)

plt.show()
```
