```python
import matplotlib.pyplot as plt
```

```python
# Write one function that predicts an output for a single number using x * weight + bias, and 
# another function that predicts outputs for every number in a list using the same formula.
```

```python
def predict_linear(x,weight,bias):
    return (x*weight)+bias 


predict_linear(5,10,2)
```

```python
def predict_many(xs,weight,bias):
    predictions = []

    for x in xs:
        predictions.append(x*weight+bias)

    return predictions

xs = [1,2,3]

predict_many(xs,10,2)
```

```python
# Test positive and negative numbers, plus an empty list. Plot predictions for at least ten x values
```

```python
assert predict_many([1,2,3],10,2) == [12,22,32]
assert predict_many([-1,-2,-3],10,2) == [-8,-18,-28]
assert predict_many([],10,2) == []
```

```python
xs = list(range(-5,6))
predictions = predict_many(xs,10,2)

plt.plot(xs,predictions, marker='o')
plt.xlabel("Values  x")
plt.ylabel("Predicitons")
plt.title("Linear Predicitons")
plt.grid(color="lightgrey")

plt.locator_params(axis='both', nbins=15)

plt.show()
```

```python
def taxi_rate(x):
    weight = 12
    bias = 40
    return (x*weight+bias)
x = 1

taxi_rate(x)
```

```python
# - **Goal:** read the direction of a relationship from data.
# - **Data:** x = [1, 2, 3, 4, 5]; y = [3, 6, 9, 12, 15].
# - **Steps:**
#     - (a) Draw a scatter plot with x on the horizontal axis and y on the vertical axis.
#     - (b) Describe the pattern: increasing, decreasing, or no clear pattern.
#     - (c) Does the graph look roughly straight or curved?
#     - (d) Write a rule that connects x and y.
# - **Check yourself:** y increases in a straight-line pattern; the rule is y = 3x.
# - **Explain:** why is a graph useful before fitting a model?
```

```python
x = [1, 2, 3, 4, 5]
y = [3, 6, 9, 12, 15]

plt.scatter(x,y,
            color="darkred")
plt.xlabel("x")
plt.ylabel("y")
plt.title("X and Y Relation")

plt.locator_params(axis="both",nbins=6)
plt.grid()

plt.show()
```
