```python
# Q1. Given data = [10, 20, 30, 40, 50], calculate the mean and the deviation of each value.
```

```python
data = [10,20,30,40,50]

mean = sum(data)/len(data)
devia = [(x-mean) for x in data]
print(devia)
```

```python
# Q10
# Given data = [5, 10, 15, 20, 25], calculate both population and sample standard deviations using Python.
```

```python
data = [5, 10, 15, 20, 25]

n = len(data)

mean = sum(data)/n

deviaiton = [(x-mean)**2 for x in data]

variance = sum(deviaiton)/n
samp_varr = sum(deviaiton)/(n-1)

popu_std_dev = variance ** 0.5
samp_std_dev = samp_varr ** 0.5

print(popu_std_dev)
print(samp_std_dev)
```
