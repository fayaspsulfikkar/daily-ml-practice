```python
# Question 1/20 🎲 — Basic Probability
# A fair six-sided die is rolled once.

# What is the probability of getting a 4?

# Think about:
# - How many possible outcomes are there?
# - How many favorable outcomes are there?
```

```python
sample_space = [1,2,3,4,5,6]
event = {4}


p_4 = len(event)/len(sample_space)

p_4 = p_4 *100

print(p_4)
```
