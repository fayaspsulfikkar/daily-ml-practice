```python
# You are given the exam scores of five students:
# scores = [45, 60, 75, 80, 90]


# Write a Python program to calculate the following:
# 1. Mean
# 2. Median
# 3. Range
# 4. Population variance
# 5. Population standard deviation
# Rules:
# - Use Python to calculate the answers.
# - Do not use NumPy, Pandas, or any other external library.
# - You may use Python's built-in functions.
```

```python
scores = [45,60,75,80,90]

mean = sum(scores)/len(scores)
print(mean)
```

```python
sorted_score = sorted(scores)
n = len(sorted_score)

if n % 2 ==1:
    median = sorted_score[n//2]
else:
    median = (sorted_score[n//2 - 1] + sorted_score[n//2]) / 2



print(median)
```

```python
score_range = max(scores) - min(scores)
print(score_range)
```

```python
diff = [(x-mean)**2 for x in scores]

varience = sum(diff) /n

print(varience)
```

```python
std_dev = varience ** 0.5 
print(std_dev)
```

```python
# Practice Question 3 — Beginner to Intermediate
# You are given the number of hours five students studied and their exam scores.
# hours = [1, 2, 3, 4, 5]scores = [30, 40, 50, 60, 70]


# Write a Python program to calculate the population covariance between study hours and exam scores.
# Requirements
# 1. Calculate the mean of hours.
# 2. Calculate the mean of scores.
# 3. Calculate the population covariance without using NumPy, Pandas, or any other external library.
# 4. Print the final covariance.
```

```python
hours = [1,2,3,4,5]
scores = [30,40,50,60,70]

mean_hours = sum(hours)/len(hours)
print("Mean of Hours: ", mean_hours)

mean_scores = sum(scores)/len(scores)
print("Mean of Scores: ",mean_scores)


hours_dev = [(x-mean_hours) for x in hours]
scores_dev = [(x-mean_scores) for x in scores]

products = [x*y for x,y in zip(hours_dev, scores_dev)]
print(products)

pop_v = sum(products) // len(hours)
print(pop_v)
```
