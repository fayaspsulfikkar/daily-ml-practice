```python
# - **Goal:** use sets of outcomes to combine events.
# - **Data:** roll a fair die. Let A = “even number” = {2, 4, 6}; B = “number greater than 3” = {4, 5, 6}.
# - **Steps:**
#     - (a) Find the complement of A: outcomes that are not in A.
#     - (b) Find A ∩ B, meaning outcomes in both A and B.
#     - (c) Find A ∪ B, meaning outcomes in A or B or both.
#     - (d) Calculate P(A ∪ B) using P(A) + P(B) - P(A ∩ B).
# - **Check yourself:** complement of A = {1, 3, 5}; A ∩ B = {4, 6}; A ∪ B = {2, 4, 5, 6}; P(A ∪ B) = 4/6 = 2/3.
# - **Explain:** why do we subtract the intersection when adding P(A) and P(B)?
```

```python
sample_space = {1,2,3,4,5,6}
event_A = {x for x in sample_space if x%2==0}
event_B = {x for x in sample_space if x>3}
print(event_A,event_B)
print(event_A|event_B)
```

```python
compl_A = sample_space - event_A
compl_A
```

```python
print(event_A & event_B)
```

```python
print(event_A | event_B)
```

```python
P_A = len(event_A) / len(sample_space)
P_B = len(event_B) / len(sample_space)
P_intersection = len(event_A & event_B) / len(sample_space)
print(P_intersection)

P_union = P_A + P_B -P_intersection
print(P_union)
```
