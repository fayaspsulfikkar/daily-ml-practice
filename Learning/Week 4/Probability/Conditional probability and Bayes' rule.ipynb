```python
# - **Goal:** calculate a conditional probability and understand Bayes' intuition.
# - **Data:** out of 1,000 people, 10 have a condition. A test detects 9 of those 10, but also gives a positive result to 99 people who do not have the condition.
# - **Steps:**
#     - (a) Count true positives.
#     - (b) Count false positives.
#     - (c) Calculate the total number of positive results.
#     - (d) Calculate the fraction of positive results that are true positives.
#     - (e) Explain why the condition's starting rate matters.
```

```python
import pandas as pd
```

```python
df = pd.read_csv("bayes_condition_test_dataset.csv", index_col="person_id")

count_tru_pos = ((df["has_condition"] == "Yes") & (df["test_result"] == "Positive")).sum()
count_fal_pos = ((df["has_condition"] == "No") & (df["test_result"] == "Positive")).sum()


total_pos = count_fal_pos + count_tru_pos
fraction = count_tru_pos / total_pos
percentage = fraction*100

hav_cond = (df["has_condition"]=="Yes").sum()
fraction_cond = hav_cond/len(df)
perecntage_cond = fraction_cond * 100

print("True positives:", count_tru_pos)
print("False positives:", count_fal_pos)
print("Total positive results:", total_pos)
print("Fraction of positive results that are true positives:", fraction)
print("Percentage of positive results that are true positives:", percentage)
print("People with the condition:", hav_cond)
print("Base rate:", fraction_cond)
print("Base rate percentage:", perecntage_cond)
```
