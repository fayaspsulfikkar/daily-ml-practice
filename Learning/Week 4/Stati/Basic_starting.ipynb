```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Calculate mean, median, and mode for 3, 4, 4, 5, 6, 8, 30. Which centre is best and why?
```

```python
array = np.array([3, 4, 4, 5, 6, 8, 30])

print("Mean:",array.mean())
print("Median:", np.median(array))
print("Mode:", pd.Series(array).mode())
```

```python
A = [8, 9, 10, 11, 12]
B = [2, 6, 10, 14, 18]

print(np.mean(A), np.std(A))
print(np.mean(B), np.std(B))
```
