```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# - **Goal:** use plots and shape descriptions to understand data.
# - **Data:** Dataset A = [2, 3, 3, 4, 5]; Dataset B = [2, 3, 3, 4, 20].
# - **Steps:**
#     - (a) Draw a histogram for each dataset.
#     - (b) Draw a box plot for each dataset.
#     - (c) Compare the means and medians.
#     - (d) Identify which dataset has a longer right tail.
#     - (e) In your own words, describe skewness and kurtosis.
# - **Check yourself:** Dataset B has a larger mean because of 20 and is more right-skewed. A box plot may flag 20 as an outlier depending on the quartile convention.
# - **Explain:** why should you use both a graph and numerical summaries?
```

```python
Dataset_A = [12, 14, 15, 15, 16, 17, 18, 18, 19, 20,
             20, 21, 22, 22, 23, 24, 25, 26, 27, 30,
             31, 32, 33, 34, 35, 36, 38, 40, 42, 45]

Dataset_B = [12, 14, 15, 15, 16, 17, 18, 18, 19, 20,
             20, 21, 22, 22, 23, 24, 25, 26, 27, 30,
             31, 32, 33, 34, 35, 36, 38, 40, 42, 100]

bins = np.linspace(10, 100, 8)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.hist(Dataset_A,bins=bins,
         rwidth=0.9)
plt.title("Dataset A")

plt.subplot(1,2,2)
plt.hist(Dataset_B,bins=bins,
         rwidth=0.9)
plt.title("Dataset B")

plt.figure(figsize=(8,8))

plt.subplot(2,2,1)
plt.boxplot(Dataset_A, 
            boxprops=dict(color="Red"),
            medianprops=dict(color="Green"),
            whiskerprops=)

plt.subplot(2,2,2)
plt.boxplot(Dataset_B,
            boxprops=dict(color="Blue"))

plt.tight_layout()

plt.show()
```

```python
print("Mean A: ", np.mean(Dataset_A))
print("Mean B: ", np.mean(Dataset_B))

print("Median A: ", np.median(Dataset_A))
print("Median B: ", np.median(Dataset_B))
```
