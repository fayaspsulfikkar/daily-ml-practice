```python
# - **Goal:** measure how spread out values are.
# - **Data:** [2, 4, 6]. Use the population variance formula for this question.
# - **Steps:**
#     - (a) Calculate the range: maximum minus minimum.
#     - (b) Calculate the mean.
#     - (c) Subtract the mean from each value and square each difference.
#     - (d) Average the squared differences to get population variance.
#     - (e) Take the square root of the variance to get population standard deviation.
# - **Check yourself:** range = 4; mean = 4; squared differences = [4, 0, 4]; variance = 8/3 ≈ 2.67; standard deviation ≈ 1.63.
# - **Explain:** what does standard deviation tell you that the range does not?
```

```python
data = [2, 4, 6]
data_range = max(data)-min(data) 
mean = sum(data)/len(data)

squared_std = [(x-mean)**2 for x in data]
variance = sum(squared_std)/len(squared_std)

std_dev = variance**0.5


print("Range:",data_range)
print("Mean: ",mean)
print("Variance: ",variance)
print("Standard Deviation: ",std_dev)
```
