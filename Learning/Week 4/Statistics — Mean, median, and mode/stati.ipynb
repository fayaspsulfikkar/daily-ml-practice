```python
# - **Goal:** calculate three common measures of the centre of data.
# - **Data:** [2, 3, 3, 4, 8].
# - **Steps:**
#     - (a) Calculate the mean by adding the values and dividing by the number of values.
#     - (b) Find the median, the middle value after sorting.
#     - (c) Find the mode, the most frequent value.
#     - (d) Explain which measure changes most if 8 is replaced with 80.
# - **Check yourself:** mean = 4; median = 3; mode = 3.
# - **Explain:** why can median be more useful than mean when data contains an extreme value?
```

```python
d = [2, 3, 3, 4, 80, 3]



mean = sum(d)/len(d)

sorted_d = sorted(d)
n = len(d)

if n%2 == 1:
    median = sorted_d[n//2]
else:
    median = (d[n//2 -1] + d[n//2]) / 2


print("Mean: ", mean)
print("Median: ",median)

mode = max(d, key=d.count)

print("Mode :",mode)
```
