```python
# A location's daily temperatures follow an approximately Normal distribution, with a mean of 28°C and a standard deviation of 3°C.

# Our goal is to generate temperature values and understand how they behave.
```

```python
temperature = np.random.normal(28,3,size=1000)
plt.hist(temperature,
         density=True,
         bins=20,
         rwidth=0.90)
plt.show()
```
