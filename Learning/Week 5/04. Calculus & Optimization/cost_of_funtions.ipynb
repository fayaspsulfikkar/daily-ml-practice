```python
import numpy as np
```

```python

x = np.array([1, 2, 3])
y = np.array([2, 4, 6])
```

```python
def cost(w, x, y):
    predictions = w * x
    errors = predictions - y
    squared_errors = errors ** 2
    m = len(y)
    final_cost = np.sum(squared_errors) / (2 * m)
    return final_cost
```

```python
print(cost(0, x, y))
print(cost(1, x, y))
print(cost(2, x, y))
print(cost(3, x, y))
```
