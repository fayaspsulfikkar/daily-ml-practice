```python
import numpy as np

A = np.array([[4, 2],
              [3, 5]])

B = np.array([[2, 1],
              [1, 1]])

C = np.array([[1, 2],
              [2, 4]])
```

```python
print(np.linalg.det(A))
print(np.linalg.det(B))
print(np.linalg.det(C))
```

```python
inv_A = np.linalg.inv(A)
print("Inverse of A: ")
print(inv_A)

inv_B = np.linalg.inv(B)
print("Inverse of B: ")
print(inv_B)
```

```python
print(A @ inv_A)
print()
print(B @ inv_B)
```
