```python
import numpy as np
```

```python
A = np.array([
    [2, 4, 1],
    [6, 3, 5],
    [1, 2, 7]
])
```

```python
A.shape
```

```python
Q, R = np.linalg.qr(A)
print(Q)
print(R)
```

```python
A_reconstructed = Q @ R
print(A_reconstructed)
```

```python
print(np.allclose(A,A_reconstructed))
```

```python
print(np.allclose(Q.T @ Q, np.eye(3)))
```

```python
A = np.array([
    [2, 4, 1],
    [6, 3, 5],
    [1, 2, 7],
    [3, 5, 2]
])
```

```python
A.shape
```

```python
U, s, Vt = np.linalg.svd(A, full_matrices=False)

print("U: \n",U)
print("s: \n",s)
print("Vt: \n",Vt)
```

```python
Sigma = np.diag(s)
Sigma
```

```python
A_reconstructed = U @ Sigma @ Vt
print(A_reconstructed)
```

```python
print(np.allclose(A,A_reconstructed))
```
