```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def calculate_slope(x1, y1, x2, y2):
    m = (y2-y1)/(x2-x1)
    return m

```

```python
print(calculate_slope(1, 3, 4, 9))
print(calculate_slope(2, 5, 6, 1))
```

```python
def fofx(x):
    y= x**2 - 4*x
    return y
```

```python
print(fofx(1))
print(fofx(2))
print(fofx(3))
print(fofx(4))
```

```python
def approximate_slope(x, h):
    y1 = fofx(x)
    y2 = fofx(x + h)

    slope = (y2 - y1) / h

    return slope
```

```python
h = 0.001

print("x = 1:", approximate_slope(1, h))
print("x = 2:", approximate_slope(2, h))
print("x = 3:", approximate_slope(3, h))
print("x = 4:", approximate_slope(4, h))
```
