```python
import numpy as np

A = np.array([
    [1, 2],
    [2, 4]
])

print(np.linalg.matrix_rank(A))
```

```python
B = np.array([
    [1, 2],
    [3, 4]
])

print(np.linalg.matrix_rank(B))
```

```python
C = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [3, 6, 9]
])

print(np.linalg.matrix_rank(C))
```

```python
A = np.array([
    [1, 2, 3],
    [0, 4, 5],
    [1, 0, 6]
])

B = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [1, 0, 1]
])


C = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

```python
A_rank = np.linalg.matrix_rank(A)
B_rank = np.linalg.matrix_rank(B)
C_rank = np.linalg.matrix_rank(C)

print(A_rank)
print(B_rank)
print(C_rank)
```

```python
A_det = np.linalg.det(A)
B_det = np.linalg.det(B)

print(A_det)
print(B_det)
```

```python
# - **Matrix A:** Rank = 3, determinant = 22. Since the determinant is nonzero, the matrix has full rank and an inverse exists.
# - **Matrix B:** Rank = 2, determinant = 0. Since the determinant is zero, the matrix does not have full rank and an inverse does not exist.
# - **Matrix C:** Rank = 2. A determinant is not defined for a non-square matrix.

# **Pattern:** For a square matrix, a nonzero determinant means full rank and an inverse exists. A zero determinant means the matrix does not have full rank and has no inverse.
```
