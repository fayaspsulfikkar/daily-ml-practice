```python
import numpy as np
```

```python
inputs = np.arange(0,6)

outputs = inputs ** 2

print(inputs)
print(outputs)
```

```python
def func(x):
    return x**2

output = func(inputs)
print(output)
```

```python
def average_rate(a, b):
    return (func(b) - func(a)) / (b - a)

print(average_rate(1, 3))
print(average_rate(2, 3))
print(average_rate(2.9, 3))
```
