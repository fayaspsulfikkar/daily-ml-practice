```python
import numpy as np 
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
x =  np.array(x)
y = np.array(y)
m = len(x)
```

```python
def cost(w):
    errors = (w * x) -y
    return np.sum(errors)/(2*m)

def gradient(w):
    errors = (w * x) - y
    return np.sum(errors *x)/m
```

```python
weights = []
costs = []
```

```python
w = 0
lr = 0.1
```

```python
for iteration in range(50):
    g = gradient(w)
    w = w - lr * g
    c = cost(w)
    weights.append(w)
    costs.append(c)     
```

```python
print("first two weights:", weights[:2])
print("first two costs:  ", costs[:2])
print("final weight:", weights[-1])
print("final cost:  ", costs[-1])
```
