```python
import numpy as np
```

```python
def cosine_sim(a,b):
    dot_p = np.dot(a,b)

    lengtha = np.linalg.norm(a)
    lengthb = np.linalg.norm(b)
    similarity = dot_p / (lengtha * lengthb)

    return similarity
```

```python
a1 = np.array([2,3])
a2 = np.array([4,5])

b1 = np.array([1,2])
b2 = np.array([-2,1])

c1 = np.array([1,2])
c2 = np.array([-3,-4])
```

```python
p1 = cosine_sim(a1,a2)
p2 = cosine_sim(b1,b2)
p3 = cosine_sim(c1,c2)
```

```python
print(p1,p2,p3, sep="\n")
```
