```python
# 🟠 Intermediate: Matrix multiplication by hand, in code · 15 min 
```

```python
A = np.array([[1, 2, 3], 
              [4, 5, 6]])

B = np.array([[7, 8], 
              [9, 10], 
              [11, 12]])
```

```python
print(B@A)
```

```python
def matmul_manual(A, B):

    rows_A = len(A)
    cols_A = len(A[0])
    cols_B = len(B[0])

   
    result = np.zeros((rows_A, cols_B), dtype=int)

    for i in range(rows_A):

        for j in range(cols_B):

            total = 0

            for k in range(cols_A):

                total += A[i][k] * B[k][j]

            result[i][j] = total

    return result


print(matmul_manual(B,A))
```

```python
a = np.array([3, 4])
b = np.array([1, 2])
```

```python
print(a.shape)
print(a.reshape(-1,1).shape)
```

```python
print("Euclidean Length: ",np.linalg.norm(a))
print("Manhattan Length: ",np.linalg.norm(a, ord=1))
print("Point Vector: ",a/np.linalg.norm(a))
print("Magnitude of Unit Vecotr a:",np.linalg.norm(a/np.linalg.norm(a)))
```

```python
print("Dot Product: ", np.dot(a,b))
```

```python
cosine = np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))

print("Cosine Similarity: ", cosine)
```

```python
print("Distance Between Euclidean and Manhattan Distance: ",np.linalg.norm(a)-np.linalg.norm(b, ord=1))
```
