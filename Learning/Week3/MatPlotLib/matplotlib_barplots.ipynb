```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Titanic-Dataset.csv")

filtered_age_class = df.dropna(subset=["Age","Pclass"])

age = filtered_age_class["Age"]
Pclass = filtered_age_class["Pclass"]

unique_class = Pclass.unique()

group1 = filtered_age_class[filtered_age_class["Pclass"]==1]["Age"]
group2 = filtered_age_class[filtered_age_class["Pclass"]==2]["Age"]
group3 = filtered_age_class[filtered_age_class["Pclass"]==3]["Age"]

age_groups = [group1,group2,group3]

plt.figure(figsize=(5,8))
plt.boxplot(age_groups)

plt.title("Age distribution by passenger class",
          fontsize=13,
          fontstyle="italic",
          color="darkblue")
plt.xlabel("Class")
plt.ylabel("Age")

plt.tight_layout()

plt.show()
```
