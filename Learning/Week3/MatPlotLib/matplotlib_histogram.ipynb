```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
scores = np.random.normal(loc=80,scale=10,size=200)
scores = np.clip(scores,1,100)

plt.hist(scores,bins=10,
         color="green",
         edgecolor="black",
         rwidth=0.8,
         )

plt.title("Exam Scores")
plt.xlabel("Marks")
plt.ylabel("# No of Students")

plt.show()
```
