```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
xaxis =  np.arange(1,151)
yaxis =  np.arange(1,151)

plt.plot(xaxis,yaxis)
plt.show()
```

```python
xaxis = np.arange(1,151)
yaxis = xaxis ** 2

plt.plot(xaxis,yaxis)
plt.show()
```

```python
xaxis = np.arange(1,101)
yaxis = xaxis * -1

plt.plot(xaxis,yaxis)
plt.show()
```

```python
xaxis = np.linspace(1,2 * np.pi,100)
yaxis = np.sin(xaxis)

plt.plot(xaxis,yaxis)
plt.show()
```

```python
xaxis = np.random.randint(1,100, size=101)
yaxis = np.random.randint(1,100, size=101)

plt.plot(xaxis,yaxis)
plt.show()
```

```python
xaxis = [2020,2021,2022,2023,2024]
yaxis = [12,20,6,34,34]

plt.plot(xaxis,yaxis)
plt.show()
```

```python
xaxis = np.arange(1,51,dtype=float)
yaxis = xaxis ** 10

plt.plot(xaxis,yaxis)
plt.show()
```
