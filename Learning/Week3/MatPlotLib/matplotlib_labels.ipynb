```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([2023,2024,2025,2026])
y1 = np.array([12,34,56,76])
y2 = np.array([2,3,1,33])
y3 = np.array([66,78,99,12])

plt.title("Sulmaan",
          fontsize=20,
          family="Serif",
          fontweight="bold",
          color="violet"
          )
plt.xlabel("Year",
           fontsize=10,
           family="Sans",
           color="purple"
           )
plt.ylabel("Rate",
            fontsize=10,
            family="Sans",
            color="purple"
            )
plt.tick_params(axis="both",
                colors="purple")


plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)

plt.xticks(x)

plt.show()
```
