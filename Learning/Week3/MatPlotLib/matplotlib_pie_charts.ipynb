```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
categories = ["Freshman","Sophomore","Junior","Senior"]
students = [100,250,500,300]
colors = ["darkred","darkgreen","#df0303","darkorange"]

textprops={
    "color": "Black",
    "fontsize": 14,
    "fontstyle": "italic",
    "fontfamily": "serif"
}

plt.title("Students Details",
          fontsize="20",
          family="Serif",
          fontstyle="italic",
          fontweight="bold",
          color="Brown")

plt.pie(students, labels=categories,
        colors=colors,
        explode=[0,0.1,0,0],
        autopct="%1.1f%%",
        shadow=True,
        startangle=320,
        textprops=textprops
)


plt.show()
```
