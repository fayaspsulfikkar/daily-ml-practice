```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Titanic-Dataset.csv")

df
```

```python
print(df["Age"].value_counts())
```

```python
age_values = df["Age"].value_counts()
age_values
```

```python
plt.bar(age_values.index, age_values.values)
plt.show()
```

```python
plt.pie(age_values.values,labels = age_values.index)
plt.show
```

```python
plt.hist(df["Age"],bins=10,edgecolor="black",rwidth=0.8)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```
