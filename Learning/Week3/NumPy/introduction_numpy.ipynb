```python
# A 2D list (3x3 grid)
matrix_list = [
 
    [7, 8, 9]
]

# Process every element using nested loops
result_list = []
for row in matrix_list:
    new_row = []
    for item in row:
        # Step-by-step math done by the slower Python interpreter
        transformed_value = (item ** 2) + 10
        new_row.append(transformed_value)
    result_list.append(new_row)

print("Python List Result:", result_list)
```

```python

import numpy as np

# A 2D NumPy array (3x3 grid)
matrix_array = np.array([
 
    [7, 8, 9]
])

# Vectorized operation: No loops required!
# Locality of reference and SIMD instructions process this instantly.
result_array = (matrix_array ** 2) + 10

print("NumPy Array Result:\n", result_array)
```
