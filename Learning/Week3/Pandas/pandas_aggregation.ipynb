```python
print(ds["Height"].sum())
print(ds["Height"].max())
```

```python
group = ds.groupby("Type1")

print(group["Height"].mean())
```

```python
group = ds.groupby("Type1")

print(group["Weight"].mean())
```

```python
height_agg = ds.groupby("Type1")["Height"].agg(["mean","median","count"])
height_agg
```

```python
height_100 = ds.groupby("Type1").filter(lambda group: (group["Height"] > 100).any())
height_100
```
