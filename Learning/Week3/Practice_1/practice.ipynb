```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1	How many passengers survived and how many died?	Choose a chart and explain why.

df = pd.read_csv("../Titanic-Dataset.csv")

survivors = df["Survived"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(survivors.index, survivors.values,
        color="Red",
        )

plt.title("Titanic Passenger Survival")
plt.xlabel("Passenger Status")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.xticks([0,1],["Died","Survived"])

plt.show()

```

```python
#  2	What was the age distribution of passengers aboard the Titanic?	Choose a chart and explain why.

ages = df["Age"]

plt.hist(ages, bins=10,
         color="Red",
         edgecolor="Black",
         rwidth=0.8)

plt.ylabel("Passengers",
           fontsize=14)
plt.xlabel("Age",
           fontsize=14)

plt.tick_params(axis="both",labelcolor="red")

plt.show
```

```python
#Was there a relationship between passenger age and fare paid

age = df["Age"]
fare = df["Fare"]

plt.scatter(age,fare,
            s=8)
plt.title("Age and Fare Relation")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show
```

```python
df_clean = df.dropna(subset=["Age","Fare"])

age = df_clean["Age"]
fare = df_clean["Fare"]

plt.scatter(age,fare,
            s=8)

plt.title("Age and Fare Relation")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()

```

```python
import matplotlib.pyplot as plt


x = ["Q1", "Q2", "Q3", "Q4"]
y1 = [15000, 18000, 14000, 22000]
y2 = [4000, 5500, 3200, 7800]


plt.plot(x, y1, label="Saless", color="#2563eb", marker="o", linewidth=2)
plt.plot(x, y2, label="Profit", color="#16a34a", marker="s", linestyle="--", linewidth=2)


plt.xlabel("Quarter")
plt.ylabel("Amount ($)")
plt.title("Quarterly Sales vs. Profit")
plt.grid(True, linestyle=":", alpha=0.6)
#plt.yscale("log")

plt.legend(loc="upper left")
plt.tight_layout()
plt.show()
```
