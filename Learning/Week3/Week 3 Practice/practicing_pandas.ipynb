```python
import pandas as pd
```

```python
df = pd.read_csv("../Titanic-Dataset.csv")
df
```

```python
#Count missing values in every column.

missing_values_count = df.isnull().sum()
missing_values_count
```

```python
missing_proportion = df.isnull().mean()

print(missing_proportion)
```

```python
df[df["Age"].isnull()].head(10)
```

```python
#Check whether complete rows are duplicated.

duplicates = df.duplicated().sum()
duplicates
```

```python
#Inspect unique values for Survived, Pclass, Sex, and Embarked.

columns = ["Survived", "Pclass", "Sex", "Embarked"]

for col in columns:
    print(col, df[col].unique())
```

```python
columns = ["Age", "Fare", "SibSp", "Parch"]

for col in columns:
    print(col, "Min:", df[col].min(), "Max:", df[col].max())
```

```python
minmax_coloumns = ["Age","Fare","SibSp","Embarked"]

for col in minmax_coloumns:
    print(f"{col} Min: {df[col].min()} Max: {df[col].max()}")
```

```python
# Produce descriptive statistics for Age, Fare, SibSp, and Parch.

print("Age:\n", df["Age"].describe())
```

```python
df[["Age","SibSp","Fare","Parch"]].describe()
```

```python
# Compare the mean and median for Age and Fare.
coloumn = ["Age","Fare"]

for col in coloumn:
    print(f"{col}: \nMean:{df[col].mean()}\nMedian:{df[col].median()}\n")
```
