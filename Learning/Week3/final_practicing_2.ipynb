```python
# ### Q7 · Mixed (NumPy + Pandas) — Array into a DataFrame (10 min)

# - **Goal:** pair NumPy values with labels and check they line up.
# - **Data:** `months = ["Jan", "Feb", "Mar", "Apr", "May"]` and `sales = np.array([120, 135, 150, 128, 160])`.
# - **Steps:**
#     - (a) Build a DataFrame with columns `month` and `sales`.
#     - (b) Add a column `double` equal to `sales * 2`, calculated from the NumPy array.
#     - (c) Print the row with the highest sales, using a condition with `sales.max()`.
#     - (d) Confirm the month in that row is May.
# - **Check yourself:** shape `(5, 3)`. `double` is `[240, 270, 300, 256, 320]`. The highest row is May with 160.
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = np.array([120, 135, 150, 128, 160])

df = pd.DataFrame(sales,index=months,columns=["Sales"])
df["Double"] = sales*2

df[df["Sales"]==df["Sales"].max()]
df[df["Sales"]==df["Sales"].max()].index[0]=="May"
```

```python
# ### Q8 · Mixed (Pandas + Matplotlib) — Count then chart (10 min)

# - **Goal:** turn a list of categories into counts and a bar chart.
# - **Data:** `groups = ["A", "B", "A", "C", "B", "A", "C", "B", "A", "C"]`.
# - **Steps:**
#     - (a) Put `groups` in a DataFrame column named `group`.
#     - (b) Count each group with `value_counts()`.
#     - (c) Draw a bar chart of the counts with a title and both axis labels.
#     - (d) Write one sentence about what the chart shows.
# - **Check yourself:** A = 4, B = 3, C = 3. The tallest bar is A.
```

```python
groups = ["A", "B", "A", "C", "B", "A", "C", "B", "A", "C"]

df = pd.DataFrame({"group": groups})
group_count = df["group"].value_counts()
print(group_count)

plt.bar(group_count.index,group_count.values)
plt.title("Members Per Group")
plt.xlabel("Group")
plt.ylabel("Count")

plt.tight_layout()

plt.show()
```

```python
# ### Q2 · NumPy — Masks, np.where and broadcasting (12 min)

# - **Goal:** change values by a condition and combine arrays of different shapes.
# - **Data:** `temps` is a 3×3 array with rows `[22, 25, 31]`, `[27, 35, 19]`, `[30, 26, 34]`.
# - **Steps:**
#     - (a) Count how many values are above 30, using a True/False mask.
#     - (b) Use `np.where` to make a copy where every value above 30 becomes 30 and all other values stay the same.
#     - (c) Subtract each row's own mean from that row, in one line, with no loops. Predict the shape of the row-mean array first.
#     - (d) Add the 1D array `[1, 2, 3]` to every row of `temps`.
# - **Check yourself:** (a) 3. (b) `[[22 25 30] [27 30 19] [30 26 30]]`. (c) row means are 26, 27, 30, so the result is `[[-4. -1. 5.] [0. 8. -8.] [0. -4. 4.]]`. (d) `[[23 27 34] [28 37 22] [31 28 37]]`.
```

```python
temps = np.array([[22, 25, 31],[27, 35, 19],[30, 26, 34]])

print((temps > 30).sum())
print(np.where(temps>30,30,temps))
print(temps - temps.mean(axis=1, keepdims=True))
new_arr = temps + np.array([1,2,3])
print(new_arr)
```
