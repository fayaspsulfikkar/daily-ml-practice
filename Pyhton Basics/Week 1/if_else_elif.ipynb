```python
#Take_three_numbers_and_print_the_largest.

num1=int(input("Enter Number 1: "))
num2=int(input("Enter Number 2: "))
num3=int(input("Enter number 3: "))

if num1>num2:
    if num1>num3:
        print("Number 1 is Greater")
    else:
        print("Number 3 is Greater")
else:
    if num2>num3:
        print("Number 2 is Greater")
    else:
        print("Number 3 is Greater")
```

```python
#0–100 units    → ₹5/unit
#101–200        → ₹7/unit
#201–300        → ₹10/unit
#301+           → ₹15/unit

unit=int(input("Enter Your Electricity Unit Usage: "))

if unit>0 and unit<=100:
    amount = unit*5
if unit>100 and unit<201:
    amount = unit*7 
if unit>200 and unit<301:
    amount =  unit*10
if unit > 300:
    amount = unit*15
if unit<0:
    print("Invalid Entry")

print(amount)
```

```python
# Student Result System

math = int(input("Enter Your Maths Mark: "))
physics = int(input("Enter Your Physics Mark: "))
python = int(input("Enter Your Python Mark: "))
att = int(input("Enter Your Attendance Percentage: "))

if math < 0 or math > 100:
    print("Invalid Mark for Math")

elif physics < 0 or physics > 100:
    print("Invalid Mark for Physics")

elif python < 0 or python > 100:
    print("Invalid Mark for Python")

elif att < 0 or att > 100:
    print("Invalid Attendance")

elif math < 40 or physics < 40 or python < 40:
    print("Failed")

elif att < 75:
    print("Failed Due to Attendance")

else:
    avg = (math + physics + python) / 3

    print("Average:", avg)

    if avg >= 90:
        print("Grade: A")
    elif avg >= 80:
        print("Grade: B")
    elif avg >= 70:
        print("Grade: C")
    elif avg >= 60:
        print("Grade: D")
    else:
        print("Grade: E")
    avg=(math+physics+python)/3
    print(avg)
```
