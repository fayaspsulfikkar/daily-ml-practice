```python
# Exercise 1: make squares with a list comprehension
squares = [number ** 2 for number in range(1, 11)]
print(squares)
```

```python
# Exercise 2: filter and transform values in one comprehension
words = ["apple", "fig", "banana", "kiwi", "cherry"]
long_words = [word.upper() for word in words if len(word) > 4]
print(long_words)
```

```python
# Exercise 3: create a dictionary with a dict comprehension
score_labels = {score: "pass" if score >= 50 else "retry" for score in [35, 50, 78, 92]}
print(score_labels)
```

```python
# Exercise 4: use a lambda for a small sorting key
students = [{"name": "Asha", "score": 88}, {"name": "Ravi", "score": 72}, {"name": "Meera", "score": 95}]
ranked_students = sorted(students, key=lambda student: student["score"], reverse=True)
print(ranked_students)
```

```python
# Exercise 5: use pass, continue, and break intentionally
for number in range(1, 10):
    if number == 2:
        pass  # Placeholder: no special work is needed yet.
    if number % 2 == 0:
        continue
    if number > 7:
        break
    print("Odd number:", number)
```
