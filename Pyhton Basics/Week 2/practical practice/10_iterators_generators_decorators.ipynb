```python
# Exercise 1: get values from an iterator one by one
colors = ["red", "green", "blue"]
color_iterator = iter(colors)
print(next(color_iterator))
print(next(color_iterator))
```

```python
# Exercise 2: create a generator for even numbers
def even_numbers(limit):
    for number in range(2, limit + 1, 2):
        yield number


print(list(even_numbers(10)))

```

```python
# Exercise 3: use a generator expression for memory-friendly processing
total = sum(number * number for number in range(1, 6))
print("Sum of squares:", total)
```

```python
# Exercise 4: add behavior around a function with a decorator
def announce(function):
    def wrapper(*args, **kwargs):
        print("Starting function")
        result = function(*args, **kwargs)
        print("Finished function")
        return result
    return wrapper


@announce
def calculate_total(prices):
    return sum(prices)


print(calculate_total([20, 35, 15]))
```

```python
# Practice challenge: make a generator that yields valid scores only
def valid_scores(values):
    for value in values:
        if isinstance(value, (int, float)) and 0 <= value <= 100:
            yield value


print(list(valid_scores([88, -1, None, 92, 105, 75])))
```
