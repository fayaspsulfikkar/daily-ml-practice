```python
# Exercise 1: test whether a number is prime
def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


print(is_prime(29), is_prime(30))
```

```python
# Exercise 2: reverse a string and check for a palindrome
def is_palindrome(text):
    cleaned_text = "".join(character.lower() for character in text if character.isalnum())
    return cleaned_text == cleaned_text[::-1]


print(is_palindrome("Malayalam"))
print(is_palindrome("Python"))
```

```python
# Exercise 3: find the second-largest unique number
numbers = [10, 5, 20, 20, 8, 15]
unique_numbers = sorted(set(numbers), reverse=True)
print("Second largest:", unique_numbers[1])
```

```python
# Exercise 4: find repeated values in a collection
values = [3, 1, 3, 2, 5, 1, 1]
seen = set()
duplicates = set()
for value in values:
    if value in seen:
        duplicates.add(value)
    seen.add(value)
print("Duplicates:", duplicates)
```

```python
# Exercise 5: turn raw records into a summary report
records = [{"city": "Kochi", "score": 80}, {"city": "Kochi", "score": 90}, {"city": "Calicut", "score": 70}]
totals = {}
counts = {}
for record in records:
    city = record["city"]
    totals[city] = totals.get(city, 0) + record["score"]
    counts[city] = counts.get(city, 0) + 1
averages = {city: totals[city] / counts[city] for city in totals}
print(averages)
```
