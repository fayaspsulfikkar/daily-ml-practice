```python
#Make a Dog class with name and age. Add a method bark() that prints "Buddy says woof!" (using that dog's name).
#Create two dogs and call bark() on each.

class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def bark(self):
        print(self.name,"says woof! am",self.age,"Years Old")

alex = Dog("Alex",3)
bruno = Dog("Bruno",5)

alex.bark()
bruno.bark()
```

```python
#Make a Rectangle class with width and height. Add area() and perimeter(). Both should return a value, not print it.

class rectangle:
    def __init__(self,widthh,heightt):
        self.widthh = widthh
        self.heightt = heightt

    def area(self):
        return self.widthh*self.heightt

    def perimeter(self):
        return 2*(self.widthh+self.heightt)

x = rectangle(5,4)

print(x.area())
print(x.perimeter())
```

```python
class Circle:
    pi = 3.14
    def __init__(self,radius):
        self.radius = radius

    def get_circumference(self):
        return 2 * Circle.pi * self.radius

circle1 = Circle(4)

print(circle1.get_circumference())
```
