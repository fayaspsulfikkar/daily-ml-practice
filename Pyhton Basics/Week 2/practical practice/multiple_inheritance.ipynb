```python
#Inheritence

class A:
    def __init__(self):
        print("Init A Calling")
    def f1(self):
        print("f1 A Prints")

class B(A):
    def __init__(self):
        super().__init__()
        print("Init B Calling")
    def f2(self):
        print("f2 B Prints")
```

```python
#Multilevel_Inheritence

class A:
    def __init__(self):
        print("Init A Calling")
    def f1(self):
        print("f1 A Prints")

class B(A):
    def __init__(self):
        #super().__init__()
        print("Init B Calling")
    def f2(self):
        print("f2 B Prints")

class C(B):
    def f3(self):
        print("f3 C Prints")

obj=C()

obj.f1()
```

```python
#Multiple_Inherintence

class A:
    def __init__(self):
        print("Init A Calling")
    def f1(self):
        print("f1 A Prints")

class B:
    def __init__(self):
        super().__init__()
        print("Init B Calling")
    def f2(self):
        print("f2 B Prints")

class C(A,B):
    def f3(self):
        print("f3 C Prints")

obj1 = C()

C.f1(obj1)
```
