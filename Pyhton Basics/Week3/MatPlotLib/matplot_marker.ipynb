```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
xaxis = np.linspace(1,100,15)
yaxis = np.sqrt(xaxis)

plt.plot(xaxis,yaxis,marker=".",
                        ms=5, #markersize
                        markerfacecolor="red", #markerfacecolour or mfc
                        markeredgecolor="red") #makrkeredgecolor or mec 
plt.show()
```

```python
np.random.seed(100)
x = np.arange(1,50,10)
y = np.random.randint(1,50,5)
y1 = np.random.randint(1,50,5)

plt.plot(x,y,
         marker=".",
         ms=5,
         mfc="Black"
         mec="Black",
         )
plt.plot(x,y1)
plt.show()
```
