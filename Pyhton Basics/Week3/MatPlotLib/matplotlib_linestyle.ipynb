```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xaxis = np.linspace(1,20,6)
yaxis = np.sqrt(xaxis)

plt.plot(xaxis,yaxis,marker=".",
                        ms=10, #markersize
                        markerfacecolor="red", #markerfacecolour or mfc
                        markeredgecolor="red",#makrkeredgecolor or mec 
                        linestyle="dotted",
                        linewidth=2,
                        color="purple") 
plt.show()
```

```python
np.random.seed(100)
x = np.arange(1,50,10)
y = np.random.randint(1,50,5)
y1 = np.random.randint(1,50,5)


linedesign1 = dict(
    marker=".",
    ms=5,
    mfc="Black",
    mec="Black",
    linestyle="dashdot",
    linewidth="1"
)

linesyle2 = dict(
    marker="p",
    ms=5,
    mfc="Red",
    mec="Red",
    linestyle="dashed",
    linewidth="1",
    color="Purple"
)

plt.plot(x,y,color="blue", **linedesign1)
plt.plot(x,y1,color="yellow", **linesyle2)
plt.show()
```
