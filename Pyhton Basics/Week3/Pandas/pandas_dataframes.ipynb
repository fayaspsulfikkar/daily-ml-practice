```python
import pandas as pd 
```

```python
data = {
    "Name": ["Spongebob","Patrick","Nelson"],
    "Age": [30,35,60]
}
```

```python
print(data)
```

```python
df = pd.DataFrame(data,index=["Employee 1","Employee 2","Employee 3"])
print(df)
print()
print(df.loc["Employee 1"])
```

```python
print(df.iloc[-1])
```

```python
#add a new coloumn
```

```python
df["job"] = ["Cook","N/A","Cashier"]

print(df)
```

```python
#add a new row
```

```python
new_row = pd.DataFrame([{"Name":"Sandy","Age":24,"job":"Doctor"}],
                       index=(["Employee 4"]))
df = pd.concat([df, new_row])

print(df)
```

```python
#add new rows
```

```python
new_rows = pd.DataFrame([{"Name":"Ayush","Age":20,"job":"Gaming"},
                         {"Name":"Fayas","Age":24,"job":"UnEmployed"}
                         ],index=(["Employee 5","Employee 6"]))

df = pd.concat([df,new_rows])
print(df)
```
