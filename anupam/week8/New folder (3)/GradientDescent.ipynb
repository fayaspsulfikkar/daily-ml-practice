Gradient Descent Implementtion

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#starting Point
x=3.0
y=4.0
```

```python
#Learning Rate 
lr=0.1
```

```python
#Number of steps
steps = 25
```

```python
#Store the path
x_vals=[x]
y_vals=[y]
```

```python
for i in range(steps):
    grad_x=2*x
    grad_y=2*y

    #update Step

    x=x-lr*grad_x
    y=y-lr*grad_y

    x_vals.append(x)
    y_vals.append(y)

print("Final Points x,y :",x,y)
```

```python
def f(x,y):
    return x**2 + y**2
```

```python
X=np.linspace(-5,5,400)
Y=np.linspace(-5,5,400)
X,Y = np.meshgrid(X,Y)
Z=f(X,Y)
```

```python
# plt.contour(X,Y,Z,levels=50,cmap='viridis')
plt.plot(x_vals,y_vals,marker='o')
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
