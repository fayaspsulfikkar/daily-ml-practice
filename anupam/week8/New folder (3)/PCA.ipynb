PCA implementation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#synthetic data 
np.random.seed(42)
x=np.random.rand(100,5)
x
```

```python
x_mean=np.mean(x)
```

```python
x_std=x-x_mean
```

```python
cov_matrix=np.cov(x_std.T)
cov_matrix
```

```python
eigenvalues,eigenvectors=np.linalg.eig(cov_matrix)
print(eigenvalues)
print(eigenvectors)
```

```python
sorted_idx=np.argsort(eigenvalues[::-1])
sorted_idx
```

```python
top_2=eigenvectors[:,sorted_idx[:2]]
top_2
```

```python
pca=x_std @ top_2
pca
```

```python
plt.scatter(pca[:,0],pca[:,1])
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x=np.random.rand(100,5)

```

```python
x_mean=np.mean(x)
```

```python
x_std=x-x_mean
```

```python
cov_matrix=np.cov(x_std.T)
```

```python
eigenvalues,eigenvectors=np.linalg.eig(cov_matrix)
print(eigenvectors)
eigenvalues
```

```python
sorted_idx=np.argsort(eigenvalues[::-1])
```

```python
top2_eigenvectors=eigenvectors[:,sorted_idx[:2]]
```

```python
PCA=x_std @ top2_eigenvectors
```

```python
plt.scatter(PCA[:,0],PCA[:,1])
plt.xlabel('X')
plt.ylabel("Y")
plt.show()
```
