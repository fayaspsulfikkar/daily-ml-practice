```python
A=np.array([
    [4,2],
    [1,3]
])
eigenvector,eigenvalue=np.linalg.eig(A)
print(A)
print(eigenvalue)
print(eigenvector)
```

```python
np.random.seed(42)
x=np.random.rand(100,5)
```

```python
x_mean=np.mean(x,axis=0)
x_mean
x_std=x-x_mean
```

```python
cov_matrix=np.cov(x_std.T)
cov_matrix
```

```python
eigenvalue1,eigenvector1=np.linalg.eig(cov_matrix)
print(eigenvalue1)
print(eigenvector1)
```

```python
sorted_idx = np.argsort(eigenvalue1)[::-1]
sorted_idx
```

```python
top_2_eigenvectors=eigenvector1[:,sorted_idx[:2]] 
x_pca = x_std @ top_2_eigenvectors
x_pca
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(x_pca[:, 0], x_pca[:, 1], c='skyblue', edgecolor='k')
plt.title("PCA Projection onto 2D")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.show()
```
