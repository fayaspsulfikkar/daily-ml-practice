```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6]
y=[6,7,8,9,4,2]
```

```python
lr=0.1
steps=20
```

```python
for x0,y0 in zip(x,y):
    x_vals=[x0]
    y_vals=[y0]
    x=x0
    y=y0

    for i in range(steps):
        grad_x=2*x
        grad_y=2*y
        x=x - lr *grad_x
        y=y-lr *grad_y
        x_vals.append(x)
        y_vals.append(y)

plt.plot(x_vals,y_vals,marker='o')
plt.xlabel('X')
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.array([1,2,3,4,5])

def f(x):
    return x**2
def grad_f(x):
    return 2*x
lr=0.1
steps=25


x_vals=[x]
y_vals=[f(x)]

for i in range(steps):
    grad_x= grad_f(x)
    x=x-lr*grad_x
    x_vals.append(x)
    y_vals.append(f(x))

xs=np.linspace(-6,6,100)
ys=f(xs)

plt.plot(xs,ys)
plt.plot(x_vals,y_vals,marker='o')
plt.show()
```
