```python
import numpy as np
```

```python
#making synthetic data

np.random.seed(42)
x=np.random.rand(100,5)
```

```python
x_mean=np.mean(x)
```

```python
x_std=x-x_mean
```

```python
cov_x=np.cov(x_std.T)
```

```python
eigenvalues,eigenvectors=np.linalg.eig(cov_x)
print(eigenvalues)
print(eigenvectors)
```

```python
sorted_idx=np.argsort(eigenvalues[::-1])
```

```python
top2_eigenvectors=eigenvectors[:,sorted_idx[:2]]
```

```python
PCA=x_std @ top2_eigenvectors
```

```python
import matplotlib.pyplot as plt

plt.scatter(PCA[:,0],PCA[:,1])
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function f(x) = x^2 and its gradient
def f(x):
    return x**2

def grad_f(x):
    return 2 * x

# Gradient descent setup
x = 5.0  # Start on the right side of the U
lr = 0.1
steps = 20

x_vals = [x]
y_vals = [f(x)]

for _ in range(steps):
    x = x - lr * grad_f(x)
    x_vals.append(x)
    y_vals.append(f(x))

# Plot the function
xs = np.linspace(-4, 4, 200)
ys = f(xs)

plt.plot(xs, ys, label='f(x) = x²')             # Plot full U-shaped parabola
plt.plot(x_vals, y_vals, marker='o', label='Descent Path')  # Descent path on curve
# plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Proper U-Shaped Gradient Descent on f(x) = x²")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
lr = 0.1
steps = 25

for x0 in x:
    x_curr = x0
    x_vals = [x_curr]
    y_vals = [x_curr**2]
    
    for i in range(steps):
        grad_x = 2 * x_curr 
        x_curr = x_curr - lr * grad_x
        x_vals.append(x_curr)
        y_vals.append(x_curr**2)
    
    plt.plot(x_vals, y_vals, marker='o', label=f'Start x={x0}')

xs = np.linspace(-6, 6, 100)
ys = xs**2
plt.plot(xs, ys, 'k-', label='y = x^2')

# plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent on y = x^2")
plt.legend()
plt.grid(True)
plt.show()
```

```python
def grad(x,lr=0.1,steps=25):
    x=np.array(x)
    x_vals=[x]

    for i in range(steps):
        grad_x=2*x
        x=x-lr*grad_x
        x_vals.append(x)
    x_vals=np.array(x_vals)
    y_vals=x_vals**2
    return x_vals,y_vals

xs=np.linspace(-6,6,100)
ys=xs**2

x=[1,2,3,4,5]
x_vals,y_vals=grad(x,lr=0.5,steps=25)

plt.plot(xs,ys)
plt.plot(x_vals,y_vals,marker='o')
plt.show()
```
