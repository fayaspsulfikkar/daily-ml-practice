```python
x=[1,2,3,4,5]
y=[2,3,4,5,8]
```

```python
lr=0.1
steps=20
```

```python
for x0,y0 in zip(x,y):
    x_vals=[x0]
    y_vals=[y0]
    x=x0
    y=y0
for i in range(steps):
    grad_x=2*x
    grad_y=2*y
    x=x-lr*grad_x
    y=y-lr*grad_y
    x_vals.append(x)
    y_vals.append(y)
```

```python
import matplotlib.pyplot as plt

plt.plot(x_vals,y_vals,marker='o')
plt.xlabel("X")
plt.ylabel("Y")
```
