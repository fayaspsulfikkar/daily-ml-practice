```python
import numpy as np
import matplotlib.pyplot as plt

# Single starting point for gradient descent
x0, y0 = 5, 8  # You can change this starting point

# Convert to NumPy arrays
x = np.array(x0, dtype=float)
y = np.array(y0, dtype=float)

lr = 0.1
steps = 50

x_vals = [x]
y_vals = [y]

# Gradient descent loop
for i in range(steps):
    grad_x = 2 * x
    grad_y = 20 * y
    x = x - lr * grad_x
    y = y - lr * grad_y
    x_vals.append(x)
    y_vals.append(y)

# Plotting
plt.plot(x_vals, y_vals, marker='o')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent Path")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Start at a point not aligned with axes
x = np.array(5.0)
y = np.array(5.0)

lr = 0.1  # Learning rate
steps = 50

x_vals = [x]
y_vals = [y]

# Gradient descent for f(x, y) = x^2 + 10*y^2
for _ in range(steps):
    grad_x = 2 * x
    grad_y = 20 * y  # Derivative of 10*y^2 is 20*y
    x = x - lr * grad_x
    y = y - lr * grad_y
    x_vals.append(x)
    y_vals.append(y)

# Plot the descent path
plt.plot(x_vals, y_vals, marker='o')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Curved Gradient Descent Path on f(x, y) = x² + 10y²")
plt.grid(True)

# Optional: draw contour lines to visualize function surface
# X, Y = np.meshgrid(np.linspace(-6, 6, 100), np.linspace(-6, 6, 100))
# Z = X**2 + 10*Y**2
# plt.contour(X, Y, Z, levels=20, alpha=0.5, cmap='viridis')

plt.show()
```
