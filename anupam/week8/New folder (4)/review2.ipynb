```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2
def grad_f(x):
    return 2*x

x=5
lr=0.1
steps=25

x_vals=[x]
y_vals=[x**2]

for i in range(steps):
    grad_x=2*x
    x=x-lr*grad_f(x)
    x_vals.append(x)
    y_vals.append(f(x))

xs=np.linspace(-6,6,100)
ys=f(xs)

plt.plot(xs,ys,label='f(x)')
plt.plot(x_vals,y_vals,marker="o",label="Descent path")
plt.legend()
plt.show()
```

```python
x=np.array([1,2,3,4,5])
lr=0.1
steps=25

x_vals=[x]
y_vals=[x**2]

for i in range(steps):
    gradx=2*x
    x=x-lr*gradx
    x_vals.append(x)
    y_vals.append(x**2)

xs=np.linspace(-6,6,100)
ys=xs**2

plt.plot(xs,ys)
plt.plot(x_vals,y_vals,marker='o')
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([1,4,9,16,25])

lr=0.5
steps=25

x_vals=[x]
y_vals=[y]

for i in range(steps):
    grad__x=2*x
    grad__y=2*y
    x=x-lr*grad__x
    y=y-lr*grad__y
    x_vals.append(x)
    y_vals.append(y)

xs=np.linspace(-6,6,100)
ys=xs**2

plt.plot(xs,ys)
plt.plot(x_vals,y_vals,marker='o')
plt.show()
```

```python
np.random.seed(42)
x=np.random.rand(100,5)
```

```python
x_mean=np.mean(x)
```

```python
x_std=x-x_mean
```

```python
cov_matrix=np.cov(x_std.T)
```

```python
eigenvalues,eigenvectors=np.linalg.eig(cov_matrix)
print(eigenvalues)
print(eigenvectors)
```

```python
sorted=np.argsort(eigenvalues[::-1])
```

```python
top2_eigenvector=eigenvectors[:,sorted[:2]]
```

```python
PCA=x_std @ top2_eigenvector
```

```python
plt.scatter(PCA[:,0],PCA[:,1])
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
