```python
import matplotlib.pyplot as plt

def f(x):
    return x**2
def grad_f(x):
    return 3*x**2

x=3
lr=0.1
steps=25

x_vals=[x]
y_vals=[f(x)]

for i in range(steps):
    grad_x=grad_f(x)
    x=x-lr*grad_f(x)
    x_vals.append(x)
    y_vals.append(f(x))

xs=np.linspace(-6,6,100)
ys=f(xs)

plt.plot(xs,ys)
plt.plot(x_vals,y_vals,marker='o')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
