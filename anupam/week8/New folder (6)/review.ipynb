```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
x=np.random.rand(100,5)
```

```python
x_mean=np.mean(x,axis=0)
```

```python
x_std=x-x_mean
```

```python
covariance=np.cov(x_std.T)
```

```python
eigenvalues,eigenvectors=np.linalg.eig(covariance)
print(eigenvalues)
print(eigenvectors)
```

```python
sorted=np.argsort(eigenvalues)[::-1]
```

```python
top2_eigenvectors=eigenvectors[:,sorted[:2]]
```

```python
PCA= x_std @ top2_eigenvectors
```

```python
plt.scatter(PCA[:,0])
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.show()
```
