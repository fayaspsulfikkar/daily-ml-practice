```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([1,2,3,4,5])
y=np.array([2,4,6,8,10])
```

```python
m=0
b=0
alpha=0.01
epochs=1000
```

```python
n=len(x)
```

```python
for i in range(epochs):
    y_pred=m*x+b
    error=y_pred-y
    m-=alpha*(2/n)*np.dot(error,x)
    b-=alpha*(2/n)*np.sum(error)

print(f"slope:{m},Intercept:{b}")
```
