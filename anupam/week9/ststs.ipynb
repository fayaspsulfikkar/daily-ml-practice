```python
import pandas as pd
import seaborn as sns
import numpy as np
from scipy import stats
```

```python
df=sns.load_dataset('titanic')
df
```

```python
df=df.dropna(subset=['age','fare'])
```

```python
#Descriptive_Statitics
numerical_Features=['age','fare']
descriptive_sts=df[numerical_Features].agg(['mean','median','std'])
descriptive_sts
```

```python
skweness=df[numerical_Features].skew()
skweness
```

```python
kurtosis=df[numerical_Features].kurtosis()
kurtosis
```

```python
#Inferential Statitics
#T_Test

survived=df[df['survived']==1]['age']
not_survived=df[df['survived']==0]['age']
t_stst,p_val=stats.ttest_ind(survived,not_survived,equal_var=False)
t_stst
p_val
```

```python
#Chi-square test
contigency=pd.crosstab(df['sex'],df['survived'])
chi2,p,dof,expected=stats.chi2_contingency(contigency)
chi2
p
dof
expected
```
