Problem Statement:

You are given a CSV file student_scores.csv with two columns:
	•	group (either “A” or “B”)
	•	score (numerical test scores)

You must:
	1.	Compute and print the mean and 95% confidence interval for the score of each group.
	2.	Perform an independent two-sample t-test to check if there’s a significant difference between the mean scores of Group A and Group B.
	3.	Interpret the p-value and print whether the null hypothesis is rejected at α = 0.05.

```python
import pandas as pd 
import numpy as np
```

```python
df= pd.read_csv('student_scores.csv')
df
```

```python
group_A=df[df['group']=='A']['score']
group_B=df[df['group']=='B']['score']
```

```python
mean_A=np.mean(group_A)
mean_b=np.mean(group_B)
```

```python
from scipy import stats
```

```python
std_error_A=stats.sem(group_A)
std_error_B=stats.sem(group_B)
```

```python
#Confidence Interval
conf_A=stats.t.interval(0.95,df=len(group_A)-1,loc=mean_A,scale=std_error_A)
conf_B=stats.t.interval(0.95,df=len(group_B)-1,loc=mean_b,scale=std_error_B)
print(conf_A)
print(conf_B)
```

```python
#ttest

t_stat,p_value=stats.ttest_ind(group_A,group_B,equal_var=False)
print(t_stat,p_value)
```

```python
a=0.05
if p_value<a:
    print("Null hypothisis rejected ")
else:
    print("failed to reject null hypothesis")
```
